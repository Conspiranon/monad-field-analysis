```python
"""
CELL 3 — FROZEN CORE ENGINE v3
===============================

This cell evaluates the frozen core's 4-component tensor field on a
periodic lattice. It produces observations, not verdicts. The engine
does not interpret ontology. The record is written to disk as text.

Reads:  nothing from Cell 2 (the translation tables are data; the
        engine does not interpret them).
Writes: observation.json, diagnostics.txt, tables_snapshot.json.

Corrections applied (2026-09-26):
  FIX 1:  I2 = Frobenius
  FIX 2:  KO is dissipative
  FIX 3:  Periodic grid, endpoint=False
  FIX 4:  Per-component operators, dPyx has + dPhi_hyb_dP_yx
  FIX 5:  I_k sourced from the frame projection, I_k = 2.01 * P
  FIX 6:  I4 = P_xx^4 + P_yy^4
  FIX 7:  I3 term dropped from the exponential
  FIX 8:  Exponential argument clipped to [-50, 0]
  FIX 9:  I1_safe floor at 1e-3
  FIX 10: dt ratios printed as observations, not CFL claims
  FIX 11: per-step diagnostics include I1, I2, I3, I4
  FIX 12: timestep described as a choice
  FIX 13: translation tables written into the ZIP
"""

import os
import shutil
import json
import numpy as np
from datetime import datetime


# =============================================================================
# ENVIRONMENT
# =============================================================================
try:
    from google.colab import drive, files
    IN_COLAB = True
    drive.mount('/content/drive')
except ImportError:
    IN_COLAB = False
    print("Not in Colab. Drive and download steps will be simulated.")

PROJECT_NAME = "FRCMD_FROZEN_CORE_V3"
timestamp    = datetime.now().strftime("%Y%m%d_%H%M%S")
output_dir   = f"output_{timestamp}"
os.makedirs(output_dir, exist_ok=True)
print(f"Workspace: {output_dir}/")


# =============================================================================
# FROZEN CORE CONSTANTS
# =============================================================================
PHI       = 1.618033988749895
PI_MAX    = 5.9259
KO_SIGMA  = 0.045

ALPHA_HYB = 1.0
BETA_HYB  = 0.1
GAMMA_HYB = 0.1
I_G       = 1.0

# Frame projection: I = Psi(Pi) = [id + 1.01 ; L_v ; (1/2)id ; (1/10)id] . Pi
# Diagonal scalar projection onto the first component:
IK_COEFF  = 1.0 + 1.01        # = 2.01


# =============================================================================
# PER-COMPONENT OPERATORS
# =============================================================================
def dPhi_hyb_dP_yx(P_yx, I1):
    """
    Exact derivative of Phi_hyb with respect to P_yx.
    Phi_hyb = alpha*P_yx + g(I1)*beta*P_yx^2 / (1 + gamma*|P_yx|)
    """
    g = I1**2 / (I1**2 + I_G**2)
    abs_p  = np.abs(P_yx)
    sign_p = np.sign(P_yx)
    num = 2.0 * P_yx * (1.0 + GAMMA_HYB * abs_p) - GAMMA_HYB * sign_p * P_yx**2
    den = (1.0 + GAMMA_HYB * abs_p) ** 2
    return ALPHA_HYB + g * BETA_HYB * num / den


def frozen_core_operator(P_xx, P_xy, P_yx, P_yy, dx):
    """
    Returns (dPxx_dt, dPxy_dt, dPyx_dt, dPyy_dt) for the frozen core v3.
    All four components are differentiated correctly.
    """
    # --- Invariants ---
    I1 = P_xx + P_yy
    I2 = P_xx**2 + P_xy**2 + P_yx**2 + P_yy**2       # Frobenius
    I4 = P_xx**4 + P_yy**4                            # frozen core I4
    I3 = P_xx * P_yy - P_xy * P_yx                    # frozen core I3 (abandoned)

    # --- I1 guard ---
    I1_safe = np.clip(I1, 1e-3, None)

    # --- Finite-response difference ---
    def D(f):
        return (np.roll(f, -1) - np.roll(f, 1)) / (2.0 * dx)

    # --- Exponential envelope: drop I3, keep I2^2 + I4^4, clip ---
    exp_arg = -0.5 * (I2**2 + I4**4)
    exp_arg = np.clip(exp_arg, -50.0, 0.0)
    envelope = np.exp(exp_arg)

    # --- C(Pi) per component ---
    def C_term(P_comp, D_P):
        I_k   = IK_COEFF * P_comp
        term1 = 0.2 * (D_P * I1)
        term2 = 0.2 * (I2 - I1) * (I1 + I2)
        term3 = 0.1 * (I_k ** 2)
        term4 = ((1.0 / PI_MAX)
                 * (I1_safe**(-0.5) - 1.0)
                 * envelope
                 * P_comp)
        return term1 + term2 + term3 + term4

    dPxx = C_term(P_xx, D(P_xx))
    dPxy = C_term(P_xy, D(P_xy))
    dPyx = C_term(P_yx, D(P_yx))
    dPyy = C_term(P_yy, D(P_yy))

    # --- Antisymmetric stress contribution to dPyx ---
    dPyx = dPyx + dPhi_hyb_dP_yx(P_yx, I1)

    # --- KO dissipation (dissipative sign) ---
    def KO(f):
        stencil = (np.roll(f, -2) - 4.0*np.roll(f, -1) + 6.0*f
                   - 4.0*np.roll(f, 1) + np.roll(f, 2))
        return KO_SIGMA * stencil / (dx**4)

    dPxx = dPxx - KO(P_xx)
    dPxy = dPxy - KO(P_xy)
    dPyx = dPyx - KO(P_yx)
    dPyy = dPyy - KO(P_yy)

    return dPxx, dPxy, dPyx, dPyy


# =============================================================================
# INTEGRATION
# =============================================================================
def run_frozen_core(N=256, L=20.0, steps=5000, dt_coeff=0.005):
    x  = np.linspace(-L/2, L/2, N, endpoint=False)
    dx = L / N
    dt = dt_coeff * (dx ** 2)

    P_xx = PHI + np.exp(-(x + 3.0)**2)
    P_yy = PHI + np.exp(-(x - 3.0)**2)
    P_xy = np.zeros(N)
    P_yx = np.zeros(N)

    history = {}
    diagnostics = []
    unstable = False
    max_update = 0.0

    print(f"Frozen core: N={N}, dx={dx:.6f}, dt={dt:.6e}, steps={steps}")
    print(f"dt / dx^2 = {dt / (dx**2):.6f}  (observation, not a CFL claim)")

    for step in range(steps + 1):
        dPxx, dPxy, dPyx, dPyy = frozen_core_operator(P_xx, P_xy, P_yx, P_yy, dx)

        step_update = max(
            float(np.max(np.abs(dt * dPxx))),
            float(np.max(np.abs(dt * dPxy))),
            float(np.max(np.abs(dt * dPyx))),
            float(np.max(np.abs(dt * dPyy))),
        )
        max_update = max(max_update, step_update)

        P_xx = P_xx + dt * dPxx
        P_xy = P_xy + dt * dPxy
        P_yx = P_yx + dt * dPyx
        P_yy = P_yy + dt * dPyy

        if step % 500 == 0:
            I1_now = P_xx + P_yy
            I2_now = P_xx**2 + P_xy**2 + P_yx**2 + P_yy**2
            I3_now = P_xx * P_yy - P_xy * P_yx
            I4_now = P_xx**4 + P_yy**4

            line = (f"Step {step:04d} | "
                    f"max|Pxx|={float(np.max(np.abs(P_xx))):.4e} | "
                    f"max|Pyx|={float(np.max(np.abs(P_yx))):.4e} | "
                    f"max|I1|={float(np.max(np.abs(I1_now))):.4e} | "
                    f"max|I2|={float(np.max(np.abs(I2_now))):.4e} | "
                    f"max|I3|={float(np.max(np.abs(I3_now))):.4e} | "
                    f"max|I4|={float(np.max(np.abs(I4_now))):.4e} | "
                    f"step_update={step_update:.4e}")
            diagnostics.append(line)
            print(line)
            history[step] = P_xx.copy()

        finite = (np.isfinite(P_xx).all() and np.isfinite(P_yy).all()
                  and np.isfinite(P_xy).all() and np.isfinite(P_yx).all())
        if not finite:
            line = f"CRITICAL: non-finite value at step {step}. Aborting."
            diagnostics.append(line)
            print(line)
            unstable = True
            break

    return x, history, diagnostics, unstable, dt, max_update


# =============================================================================
# RUN
# =============================================================================
x_grid, evolution, diags, unstable, dt, max_update = run_frozen_core()


# =============================================================================
# OBSERVATION RECORD
# =============================================================================
obs_path = os.path.join(output_dir, "observation.json")
with open(obs_path, "w") as f:
    json.dump({
        "engine":          "frozen_core_v3",
        "N":               int(len(x_grid)),
        "steps_run":       len(diags),
        "dt":              float(dt),
        "max_step_update": float(max_update),
        "unstable":        bool(unstable),
        "diagnostics":     diags,
        "status":          "OBSERVATION_ONLY",
        "timestamp":       datetime.now().isoformat(),
    }, f, indent=2)

report_path = os.path.join(output_dir, "diagnostics.txt")
with open(report_path, "w") as f:
    f.write("FRCMD FROZEN CORE v3 — OBSERVATION LOG\n")
    f.write("======================================\n\n")
    for line in diags:
        f.write(line + "\n")
    f.write("\n")
    if unstable:
        f.write("STATUS: non-finite value encountered. Run aborted.\n")
    else:
        f.write("STATUS: all scheduled steps completed. No non-finite values.\n")
    f.write("This is an observation. Not a verdict on the axiom.\n")


# =============================================================================
# TABLES SNAPSHOT
# =============================================================================
TABLES_SNAPSHOT = {}
for name in ["ONTOLOGY_TRANSLATION", "CODE_PURGE", "SYMBOL_PURGE",
             "TORSION_GLOSSARY", "MECH_ANALOGY", "INVARIANTS",
             "OPERATORS", "PURGE_MAP", "FORCE_GLOSSARY",
             "TIME_AND_PRESENT_GLOSSARY", "PIVOT_GLOSSARY",
             "REGISTER_TAGS", "DIV_FR_GLOSSARY", "C_PI_GLOSSARY"]:
    if name in globals():
        TABLES_SNAPSHOT[name] = globals()[name]

if TABLES_SNAPSHOT:
    tables_path = os.path.join(output_dir, "tables_snapshot.json")
    with open(tables_path, "w", encoding="utf-8") as f:
        json.dump(TABLES_SNAPSHOT, f, indent=2, ensure_ascii=False)
    print(f"Tables snapshot written: {len(TABLES_SNAPSHOT)} tables")
else:
    print("Cell 2 tables not in namespace. Snapshot skipped.")


# =============================================================================
# PRESERVATION
# =============================================================================
zip_base = f"{PROJECT_NAME}_{timestamp}"
zip_path = shutil.make_archive(zip_base, 'zip', output_dir)

drive_ok = False
if IN_COLAB:
    try:
        drive_dir = f"/content/drive/MyDrive/{PROJECT_NAME}/{output_dir}"
        os.makedirs(drive_dir, exist_ok=True)
        for item in os.listdir(output_dir):
            shutil.copy2(os.path.join(output_dir, item),
                         os.path.join(drive_dir, item))
        shutil.copy2(zip_path, f"/content/drive/MyDrive/{PROJECT_NAME}/{zip_base}.zip")
        drive_ok = True
    except Exception as e:
        print(f"Drive backup failed: {e}")

download_ok = False
if IN_COLAB:
    try:
        files.download(zip_path)
        download_ok = True
    except Exception as e:
        print(f"Download failed: {e}")


# =============================================================================
# BENCH READOUT
# =============================================================================
print()
print("=" * 70)
print(" FRCMD FROZEN CORE v3 — BENCH READOUT")
print("=" * 70)
print(f" Output directory  : {os.path.abspath(output_dir)}")
print(f" ZIP archive       : {os.path.abspath(zip_path)}")
print(f" Drive backup      : {drive_ok}")
print(f" Download          : {download_ok}")
print(f" Steps completed   : {len(diags)}")
print(f" Unstable          : {unstable}")
print(f" dt                : {dt:.6e}")
print(f" max step update   : {max_update:.6e}")
print(f" Files in output   : {len(os.listdir(output_dir))}")
print(f" Archive size      : {os.path.getsize(zip_path)} bytes")
print("=" * 70)
```

```python
"""
CELL 4 — HOOKE / DUFFING / PIVOT TEST v2
=========================================

Checks:
  1. Linear Hooke: +kx + (-kx) = 0
  2. Duffing spring: F = -beta*x - gamma*x^3
  3. Pivot: x_pivot = sqrt(beta/gamma) where |beta*x| == |gamma*x^3|
  4. Signed balance across the pivot, with a branch for the exact tie
  5. Linear two-sided magnitudes: |F_applied| vs |F_restoring|
  6. Regime split: fraction of force from the cubic

No ontology. No verdict. Numbers only.
"""

import numpy as np

beta  = 1.0
gamma = 1.0
k     = 1.0

# --- TEST 1 ---
print("=" * 60)
print("TEST 1 — Linear Hooke: +kx + (-kx)")
print("=" * 60)
for x in [0.0, 0.5, 1.0, 2.0, 5.0]:
    F_app = +k * x
    F_res = -k * x
    net   = F_app + F_res
    print(f"  x={x:6.2f}  F_app={F_app:+8.4f}  F_res={F_res:+8.4f}  net={net:+.3e}")

# --- TEST 2 ---
print()
print("=" * 60)
print("TEST 2 — Duffing spring: F = -beta*x - gamma*x^3")
print("=" * 60)
for x in [0.1, 0.5, 1.0, 2.0, 5.0]:
    F_lin = -beta * x
    F_cub = -gamma * x**3
    F_tot = F_lin + F_cub
    ratio = abs(F_cub / F_lin) if F_lin != 0 else np.inf
    print(f"  x={x:5.2f}  F_lin={F_lin:+10.4f}  F_cub={F_cub:+10.4f}  "
          f"F_tot={F_tot:+10.4f}  |cub/lin|={ratio:.4f}")

# --- TEST 3 ---
print()
print("=" * 60)
print("TEST 3 — Pivot: |beta*x| == |gamma*x^3|")
print("=" * 60)
x_pivot_analytic = np.sqrt(beta / gamma)
print(f"  Analytic x_pivot = sqrt(beta/gamma) = {x_pivot_analytic:.6f}")
x = x_pivot_analytic
print(f"  |beta*x|      = {abs(beta * x):.6f}")
print(f"  |gamma*x^3|   = {abs(gamma * x**3):.6f}")
print(f"  difference    = {abs(beta * x) - abs(gamma * x**3):.3e}")

# --- TEST 4, corrected with tie branch ---
print()
print("=" * 60)
print("TEST 4 — Signed balance across the pivot (with tie branch)")
print("=" * 60)
for x in [0.5, 1.0, 1.5]:
    F_lin = -beta * x
    F_cub = -gamma * x**3
    if abs(F_cub) > abs(F_lin):
        status = "cubic dominates"
    elif abs(F_cub) < abs(F_lin):
        status = "linear dominates"
    else:
        status = "perfectly balanced at pivot"
    print(f"  x={x:5.2f}  F_lin={F_lin:+8.4f}  F_cub={F_cub:+8.4f}  {status}")

# --- TEST 5, relabeled ---
print()
print("=" * 60)
print("TEST 5 — Linear two-sided magnitudes: |F_applied| vs |F_restoring|")
print("       (linear Hooke only; not the Duffing total)")
print("=" * 60)
for x in [0.5, 1.0, 2.0]:
    F_app = +k * x
    F_res = -k * x
    print(f"  x={x:5.2f}  |F_app|={abs(F_app):.6f}  |F_res|={abs(F_res):.6f}  "
          f"difference={abs(F_app) - abs(F_res):.3e}")

# --- TEST 5b ---
print()
print("=" * 60)
print("TEST 5b — Duffing total magnitudes for comparison")
print("=" * 60)
for x in [0.5, 1.0, 2.0]:
    F_tot_abs = abs(-beta * x - gamma * x**3)
    print(f"  x={x:5.2f}  |F_tot| = {F_tot_abs:.6f}")

# --- TEST 6 ---
print()
print("=" * 60)
print("TEST 6 — Regime split at the pivot")
print("=" * 60)
x_p = np.sqrt(beta / gamma)
for x in [0.5 * x_p, x_p, 2.0 * x_p]:
    F_lin = -beta * x
    F_cub = -gamma * x**3
    frac_cub = abs(F_cub) / (abs(F_lin) + abs(F_cub))
    print(f"  x={x:6.4f}  fraction of force from cubic = {frac_cub:.4f}")

print()
print("=" * 60)
print("DONE. All numbers printed. No verdict.")
print("=" * 60)
```

```python
"""
I_k CONFIGURATION TEST — SCALAR vs VECTOR
==========================================

Question: is I_k in 0.1 * I_k^2 the scalar first frame component
          (I_k = 2.01 * P), or the vector norm over all four frame
          components (I_k^2 = sum of squares)?

Test:     run the v3 engine twice.
          Run A: I_k = 2.01 * P_comp                 (scalar reading)
          Run B: I_k^2 = sum of squares of frame      (vector reading)
                 with L_v component set to zero
                 (SPINE 15/16 open, so L_v is unavailable)

Compare:  does Run A diverge at the same step as Run B?
          which term dominates the divergence?
          do I1, I2, I3, I4 grow at the same rates?

No ontology. No verdict. Numbers only.
"""

import numpy as np


# =============================================================================
# CONSTANTS (same as Cell 3)
# =============================================================================
PHI      = 1.618033988749895
PI_MAX   = 5.9259
KO_SIGMA = 0.045

ALPHA_HYB = 1.0
BETA_HYB  = 0.1
GAMMA_HYB = 0.1
I_G       = 1.0

IK_COEFF  = 1.0 + 1.01     # = 2.01
HALF_ID   = 0.5            # (1/2) * id
TENTH_ID  = 0.1            # (1/10) * id


# =============================================================================
# SHARED PIECES
# =============================================================================
def dPhi_hyb_dP_yx(P_yx, I1):
    g = I1**2 / (I1**2 + I_G**2)
    abs_p  = np.abs(P_yx)
    sign_p = np.sign(P_yx)
    num = 2.0 * P_yx * (1.0 + GAMMA_HYB * abs_p) - GAMMA_HYB * sign_p * P_yx**2
    den = (1.0 + GAMMA_HYB * abs_p) ** 2
    return ALPHA_HYB + g * BETA_HYB * num / den


def _shared_invariants(P_xx, P_xy, P_yx, P_yy):
    I1 = P_xx + P_yy
    I2 = P_xx**2 + P_xy**2 + P_yx**2 + P_yy**2
    I4 = P_xx**4 + P_yy**4
    return I1, I2, I4


def _envelope(I2, I4):
    exp_arg = -0.5 * (I2**2 + I4**4)
    exp_arg = np.clip(exp_arg, -50.0, 0.0)
    return np.exp(exp_arg)


def _KO(f, dx):
    stencil = (np.roll(f, -2) - 4.0*np.roll(f, -1) + 6.0*f
               - 4.0*np.roll(f, 1) + np.roll(f, 2))
    return KO_SIGMA * stencil / (dx**4)


# =============================================================================
# READING A — SCALAR:  I_k = 2.01 * P_comp
# =============================================================================
def operator_scalar(P_xx, P_xy, P_yx, P_yy, dx):
    I1, I2, I4 = _shared_invariants(P_xx, P_xy, P_yx, P_yy)
    I1_safe = np.clip(I1, 1e-3, None)
    env     = _envelope(I2, I4)

    def D(f):
        return (np.roll(f, -1) - np.roll(f, 1)) / (2.0 * dx)

    def C_term(P_comp, D_P):
        I_k   = IK_COEFF * P_comp
        term1 = 0.2 * (D_P * I1)
        term2 = 0.2 * (I2 - I1) * (I1 + I2)
        term3 = 0.1 * (I_k ** 2)
        term4 = ((1.0 / PI_MAX)
                 * (I1_safe**(-0.5) - 1.0)
                 * env
                 * P_comp)
        return term1 + term2 + term3 + term4, (term1, term2, term3, term4)

    dPxx, t_xx = C_term(P_xx, D(P_xx))
    dPxy, t_xy = C_term(P_xy, D(P_xy))
    dPyx, t_yx = C_term(P_yx, D(P_yx))
    dPyy, t_yy = C_term(P_yy, D(P_yy))

    dPyx = dPyx + dPhi_hyb_dP_yx(P_yx, I1)

    dPxx = dPxx - _KO(P_xx, dx)
    dPxy = dPxy - _KO(P_xy, dx)
    dPyx = dPyx - _KO(P_yx, dx)
    dPyy = dPyy - _KO(P_yy, dx)

    return dPxx, dPxy, dPyx, dPyy, t_xx


# =============================================================================
# READING B — VECTOR:  I_k^2 = sum of squares of frame components
# =============================================================================
def operator_vector(P_xx, P_xy, P_yx, P_yy, dx, Lv_component=0.0):
    """
    Frame: I = Psi(Pi) = [ (id + 1.01), L_v, (1/2)id, (1/10)id ] . Pi
    I_k^2 = (2.01*P)^2 + (L_v * P)^2 + (0.5*P)^2 + (0.1*P)^2
    L_v is unavailable (SPINE 15/16). Set to zero for this test.
    """
    I1, I2, I4 = _shared_invariants(P_xx, P_xy, P_yx, P_yy)
    I1_safe = np.clip(I1, 1e-3, None)
    env     = _envelope(I2, I4)

    # Vector I_k^2 per component. L_v = Lv_component (default 0.0).
    def I_k_sq(P_comp):
        a = IK_COEFF * P_comp
        b = Lv_component * P_comp
        c = HALF_ID     * P_comp
        d = TENTH_ID    * P_comp
        return a**2 + b**2 + c**2 + d**2

    def D(f):
        return (np.roll(f, -1) - np.roll(f, 1)) / (2.0 * dx)

    def C_term(P_comp, D_P):
        term1 = 0.2 * (D_P * I1)
        term2 = 0.2 * (I2 - I1) * (I1 + I2)
        term3 = 0.1 * I_k_sq(P_comp)
        term4 = ((1.0 / PI_MAX)
                 * (I1_safe**(-0.5) - 1.0)
                 * env
                 * P_comp)
        return term1 + term2 + term3 + term4, (term1, term2, term3, term4)

    dPxx, t_xx = C_term(P_xx, D(P_xx))
    dPxy, t_xy = C_term(P_xy, D(P_xy))
    dPyx, t_yx = C_term(P_yx, D(P_yx))
    dPyy, t_yy = C_term(P_yy, D(P_yy))

    dPyx = dPyx + dPhi_hyb_dP_yx(P_yx, I1)

    dPxx = dPxx - _KO(P_xx, dx)
    dPxy = dPxy - _KO(P_xy, dx)
    dPyx = dPyx - _KO(P_yx, dx)
    dPyy = dPyy - _KO(P_yy, dx)

    return dPxx, dPxy, dPyx, dPyy, t_xx


# =============================================================================
# INTEGRATOR (shared)
# =============================================================================
def run(operator, N=256, L=20.0, steps=5000, dt_coeff=0.005, label=""):
    x  = np.linspace(-L/2, L/2, N, endpoint=False)
    dx = L / N
    dt = dt_coeff * (dx ** 2)

    P_xx = PHI + np.exp(-(x + 3.0)**2)
    P_yy = PHI + np.exp(-(x - 3.0)**2)
    P_xy = np.zeros(N)
    P_yx = np.zeros(N)

    print("=" * 74)
    print(f" RUN: {label}")
    print("=" * 74)
    print(f" N={N}  dx={dx:.6f}  dt={dt:.6e}  steps={steps}")

    for step in range(steps + 1):
        dPxx, dPxy, dPyx, dPyy, terms = operator(P_xx, P_xy, P_yx, P_yy, dx)

        P_xx = P_xx + dt * dPxx
        P_xy = P_xy + dt * dPxy
        P_yx = P_yx + dt * dPyx
        P_yy = P_yy + dt * dPyy

        if step % 500 == 0:
            I1 = P_xx + P_yy
            I2 = P_xx**2 + P_xy**2 + P_yx**2 + P_yy**2
            I3 = P_xx * P_yy - P_xy * P_yx
            I4 = P_xx**4 + P_yy**4
            t1, t2, t3, t4 = terms
            line = (f"Step {step:04d} | "
                    f"max|Pxx|={float(np.max(np.abs(P_xx))):.4e} | "
                    f"max|I2|={float(np.max(np.abs(I2))):.4e} | "
                    f"max|I4|={float(np.max(np.abs(I4))):.4e}")
            print(line)
            # term magnitudes for P_xx at this step
            tm = (f"         term1={float(np.max(np.abs(t1))):.4e} | "
                  f"term2={float(np.max(np.abs(t2))):.4e} | "
                  f"term3={float(np.max(np.abs(t3))):.4e} | "
                  f"term4={float(np.max(np.abs(t4))):.4e}")
            print(tm)

        finite = (np.isfinite(P_xx).all() and np.isfinite(P_yy).all()
                  and np.isfinite(P_xy).all() and np.isfinite(P_yx).all())
        if not finite:
            print(f"CRITICAL: non-finite value at step {step}. Aborting.")
            return {"label": label, "aborted_at": step, "status": "non_finite"}

    print(f"Reached step {steps} without non-finite values.")
    return {"label": label, "aborted_at": None, "status": "stable"}


# =============================================================================
# RUN BOTH
# =============================================================================
result_A = run(operator_scalar, label="READING A: I_k = 2.01 * P (scalar)")
result_B = run(operator_vector, label="READING B: I_k^2 = sum of squares (vector, L_v=0)")

print()
print("=" * 74)
print(" COMPARISON")
print("=" * 74)
print(f" Run A (scalar): aborted at step {result_A['aborted_at']}  status={result_A['status']}")
print(f" Run B (vector): aborted at step {result_B['aborted_at']}  status={result_B['status']}")
print("=" * 74)
```

```python
"""
ASYMPTOTE TEST — CONVERGE vs DIVERGE
=====================================

Question: is the asymptote convergent or divergent?
          does the operator have a stable fixed point?

Configurations:
  A: v4 as written              (saturating denom, thresholds = PI_MAX)
  B: v4 with higher thresholds  (I2_PIVOT = 100, IK_PIVOT = 100)
  C: v3 (no saturation)         (baseline)
  D: v4 + fixed-point pull      (explicit attraction to the pivot)

Same initial state. Same constants. Only the operator differs.

No ontology. No verdict. Numbers only.
"""

import numpy as np


PHI       = 1.618033988749895
PI_MAX    = 5.9259
KO_SIGMA  = 0.045

ALPHA_HYB = 1.0
BETA_HYB  = 0.1
GAMMA_HYB = 0.1
I_G       = 1.0

IK_COEFF  = 1.0 + 1.01        # = 2.01


def dPhi_hyb_dP_yx(P_yx, I1):
    g = I1**2 / (I1**2 + I_G**2)
    abs_p  = np.abs(P_yx)
    sign_p = np.sign(P_yx)
    num = 2.0 * P_yx * (1.0 + GAMMA_HYB * abs_p) - GAMMA_HYB * sign_p * P_yx**2
    den = (1.0 + GAMMA_HYB * abs_p) ** 2
    return ALPHA_HYB + g * BETA_HYB * num / den


def _KO(f, dx):
    stencil = (np.roll(f, -2) - 4.0*np.roll(f, -1) + 6.0*f
               - 4.0*np.roll(f, 1) + np.roll(f, 2))
    return KO_SIGMA * stencil / (dx**4)


# =============================================================================
# CONFIG A — v4 as written
# =============================================================================
def operator_A(P_xx, P_xy, P_yx, P_yy, dx,
               I2_PIVOT=PI_MAX, IK_PIVOT=PI_MAX):
    I1 = P_xx + P_yy
    I2 = P_xx**2 + P_xy**2 + P_yx**2 + P_yy**2
    I4 = P_xx**4 + P_yy**4
    I1_safe = np.clip(I1, 1e-3, None)

    def D(f):
        return (np.roll(f, -1) - np.roll(f, 1)) / (2.0 * dx)

    exp_arg = -0.5 * (I2**2 + I4**4)
    exp_arg = np.clip(exp_arg, -50.0, 0.0)
    envelope = np.exp(exp_arg)

    sat2 = 1.0 + np.maximum(I2, 0.0) / I2_PIVOT

    def C_term(P_comp, D_P):
        I_k = IK_COEFF * P_comp
        I_k_sq = I_k ** 2
        sat3 = 1.0 + I_k_sq / (IK_PIVOT ** 2)
        term1 = 0.2 * (D_P * I1)
        term2 = 0.2 * (I2 - I1) * (I1 + I2) / sat2
        term3 = 0.1 * I_k_sq / sat3
        term4 = ((1.0 / PI_MAX) * (I1_safe**(-0.5) - 1.0)
                 * envelope * P_comp)
        return term1 + term2 + term3 + term4

    dPxx = C_term(P_xx, D(P_xx))
    dPxy = C_term(P_xy, D(P_xy))
    dPyx = C_term(P_yx, D(P_yx))
    dPyy = C_term(P_yy, D(P_yy))
    dPyx = dPyx + dPhi_hyb_dP_yx(P_yx, I1)

    return (dPxx - _KO(P_xx, dx),
            dPxy - _KO(P_xy, dx),
            dPyx - _KO(P_yx, dx),
            dPyy - _KO(P_yy, dx))


# =============================================================================
# CONFIG B — v4 with higher thresholds
# =============================================================================
def operator_B(P_xx, P_xy, P_yx, P_yy, dx):
    return operator_A(P_xx, P_xy, P_yx, P_yy, dx,
                      I2_PIVOT=100.0, IK_PIVOT=100.0)


# =============================================================================
# CONFIG C — v3 (no saturation)
# =============================================================================
def operator_C(P_xx, P_xy, P_yx, P_yy, dx):
    I1 = P_xx + P_yy
    I2 = P_xx**2 + P_xy**2 + P_yx**2 + P_yy**2
    I4 = P_xx**4 + P_yy**4
    I1_safe = np.clip(I1, 1e-3, None)

    def D(f):
        return (np.roll(f, -1) - np.roll(f, 1)) / (2.0 * dx)

    exp_arg = -0.5 * (I2**2 + I4**4)
    exp_arg = np.clip(exp_arg, -50.0, 0.0)
    envelope = np.exp(exp_arg)

    def C_term(P_comp, D_P):
        I_k = IK_COEFF * P_comp
        term1 = 0.2 * (D_P * I1)
        term2 = 0.2 * (I2 - I1) * (I1 + I2)
        term3 = 0.1 * (I_k ** 2)
        term4 = ((1.0 / PI_MAX) * (I1_safe**(-0.5) - 1.0)
                 * envelope * P_comp)
        return term1 + term2 + term3 + term4

    dPxx = C_term(P_xx, D(P_xx))
    dPxy = C_term(P_xy, D(P_xy))
    dPyx = C_term(P_yx, D(P_yx))
    dPyy = C_term(P_yy, D(P_yy))
    dPyx = dPyx + dPhi_hyb_dP_yx(P_yx, I1)

    return (dPxx - _KO(P_xx, dx),
            dPxy - _KO(P_xy, dx),
            dPyx - _KO(P_yx, dx),
            dPyy - _KO(P_yy, dx))


# =============================================================================
# CONFIG D — v4 + explicit fixed-point pull toward the pivot
# =============================================================================
# The pull is a linear restoring term: kappa_pull * (P_target - P_comp)
# It attracts each component toward the pivot value PHI.
# If the response converges, the pull is doing its job.
def operator_D(P_xx, P_xy, P_yx, P_yy, dx,
               I2_PIVOT=PI_MAX, IK_PIVOT=PI_MAX, kappa_pull=0.1):
    dPxx_A, dPxy_A, dPyx_A, dPyy_A = operator_A(
        P_xx, P_xy, P_yx, P_yy, dx, I2_PIVOT, IK_PIVOT
    )
    dPxx = dPxx_A + kappa_pull * (PHI - P_xx)
    dPxy = dPxy_A + kappa_pull * (0.0 - P_xy)
    dPyx = dPyx_A + kappa_pull * (0.0 - P_yx)
    dPyy = dPyy_A + kappa_pull * (PHI - P_yy)
    return dPxx, dPxy, dPyx, dPyy


# =============================================================================
# INTEGRATOR
# =============================================================================
def run(operator, label, N=256, L=20.0, steps=2000, dt_coeff=0.005):
    x  = np.linspace(-L/2, L/2, N, endpoint=False)
    dx = L / N
    dt = dt_coeff * (dx ** 2)

    P_xx = PHI + np.exp(-(x + 3.0)**2)
    P_yy = PHI + np.exp(-(x - 3.0)**2)
    P_xy = np.zeros(N)
    P_yx = np.zeros(N)

    print("=" * 74)
    print(f" RUN: {label}")
    print("=" * 74)

    for step in range(steps + 1):
        dPxx, dPxy, dPyx, dPyy = operator(P_xx, P_xy, P_yx, P_yy, dx)

        P_xx = P_xx + dt * dPxx
        P_xy = P_xy + dt * dPxy
        P_yx = P_yx + dt * dPyx
        P_yy = P_yy + dt * dPyy

        if step % 250 == 0:
            I1 = P_xx + P_yy
            I2 = P_xx**2 + P_xy**2 + P_yx**2 + P_yy**2
            I4 = P_xx**4 + P_yy**4
            print(f"  Step {step:04d} | max|Pxx|={float(np.max(np.abs(P_xx))):.4e} | "
                  f"max|I2|={float(np.max(np.abs(I2))):.4e} | "
                  f"max|I4|={float(np.max(np.abs(I4))):.4e}")

        finite = (np.isfinite(P_xx).all() and np.isfinite(P_yy).all()
                  and np.isfinite(P_xy).all() and np.isfinite(P_yx).all())
        if not finite:
            print(f"  ABORT: non-finite value at step {step}.")
            return {"label": label, "aborted_at": step}

    print(f"  Reached step {steps} with finite values.")
    return {"label": label, "aborted_at": None}


# =============================================================================
# RUN ALL FOUR
# =============================================================================
print()
print("#" * 74)
print("# CONFIG A — v4 as written (I2_PIVOT = IK_PIVOT = PI_MAX)")
print("#" * 74)
result_A = run(operator_A, "A: v4, thresholds = PI_MAX")

print()
print("#" * 74)
print("# CONFIG B — v4 with higher thresholds (100, 100)")
print("#" * 74)
result_B = run(operator_B, "B: v4, thresholds = 100")

print()
print("#" * 74)
print("# CONFIG C — v3 (no saturation)")
print("#" * 74)
result_C = run(operator_C, "C: v3, no saturation")

print()
print("#" * 74)
print("# CONFIG D — v4 + fixed-point pull toward PHI")
print("#" * 74)
result_D = run(operator_D, "D: v4 + pull toward PHI")


print()
print("=" * 74)
print(" COMPARISON")
print("=" * 74)
for r in [result_A, result_B, result_C, result_D]:
    ab = r["aborted_at"]
    print(f"  {r['label']:35s}  aborted_at = {ab}")
print("=" * 74)
```

```python
"""
SOLITON / COLLAPSE / UNKNOT — THREE TESTS
==========================================

Test 1 — Peak width over time
    Width shrinks  -> collapse
    Width constant -> soliton
    Width grows    -> dispersion

Test 2 — Topological invariant I3 over time
    I3 monotonic   -> topology changing
    I3 oscillating -> topology preserved

Test 3 — Long run to find a release
    Sudden drop     -> unknotting event
    Continued growth -> heading toward collapse

Same v4 operator throughout. Diagnostics only.
"""

import numpy as np


PHI       = 1.618033988749895
PI_MAX    = 5.9259
KO_SIGMA  = 0.045

ALPHA_HYB = 1.0
BETA_HYB  = 0.1
GAMMA_HYB = 0.1
I_G       = 1.0

IK_COEFF  = 1.0 + 1.01
I2_PIVOT  = PI_MAX
IK_PIVOT  = PI_MAX


def dPhi_hyb_dP_yx(P_yx, I1):
    g = I1**2 / (I1**2 + I_G**2)
    abs_p  = np.abs(P_yx)
    sign_p = np.sign(P_yx)
    num = 2.0 * P_yx * (1.0 + GAMMA_HYB * abs_p) - GAMMA_HYB * sign_p * P_yx**2
    den = (1.0 + GAMMA_HYB * abs_p) ** 2
    return ALPHA_HYB + g * BETA_HYB * num / den


def _KO(f, dx):
    stencil = (np.roll(f, -2) - 4.0*np.roll(f, -1) + 6.0*f
               - 4.0*np.roll(f, 1) + np.roll(f, 2))
    return KO_SIGMA * stencil / (dx**4)


def operator_v4(P_xx, P_xy, P_yx, P_yy, dx):
    I1 = P_xx + P_yy
    I2 = P_xx**2 + P_xy**2 + P_yx**2 + P_yy**2
    I4 = P_xx**4 + P_yy**4
    I1_safe = np.clip(I1, 1e-3, None)

    def D(f):
        return (np.roll(f, -1) - np.roll(f, 1)) / (2.0 * dx)

    exp_arg = -0.5 * (I2**2 + I4**4)
    exp_arg = np.clip(exp_arg, -50.0, 0.0)
    envelope = np.exp(exp_arg)

    sat2 = 1.0 + np.maximum(I2, 0.0) / I2_PIVOT

    def C_term(P_comp, D_P):
        I_k = IK_COEFF * P_comp
        I_k_sq = I_k ** 2
        sat3 = 1.0 + I_k_sq / (IK_PIVOT ** 2)
        term1 = 0.2 * (D_P * I1)
        term2 = 0.2 * (I2 - I1) * (I1 + I2) / sat2
        term3 = 0.1 * I_k_sq / sat3
        term4 = ((1.0 / PI_MAX) * (I1_safe**(-0.5) - 1.0)
                 * envelope * P_comp)
        return term1 + term2 + term3 + term4

    dPxx = C_term(P_xx, D(P_xx))
    dPxy = C_term(P_xy, D(P_xy))
    dPyx = C_term(P_yx, D(P_yx))
    dPyy = C_term(P_yy, D(P_yy))
    dPyx = dPyx + dPhi_hyb_dP_yx(P_yx, I1)

    return (dPxx - _KO(P_xx, dx),
            dPxy - _KO(P_xy, dx),
            dPyx - _KO(P_yx, dx),
            dPyy - _KO(P_yy, dx))


# =============================================================================
# WIDTH MEASUREMENT — FWHM of the primary peak
# =============================================================================
def fwhm(profile, x):
    """
    Full width at half maximum of the primary peak.
    Returns None if no clear peak is found.
    """
    prof = np.abs(profile)
    max_idx = int(np.argmax(prof))
    max_val = prof[max_idx]
    if max_val <= 0:
        return None
    half = 0.5 * max_val

    # walk left
    left = max_idx
    while left > 0 and prof[left] > half:
        left -= 1
    # walk right
    right = max_idx
    while right < len(prof) - 1 and prof[right] > half:
        right += 1

    # interpolate between indices
    width = x[right] - x[left]
    return width


# =============================================================================
# RUN WITH DIAGNOSTICS
# =============================================================================
def run(steps=20000, N=256, L=20.0, dt_coeff=0.005, sample_every=1000):
    x  = np.linspace(-L/2, L/2, N, endpoint=False)
    dx = L / N
    dt = dt_coeff * (dx ** 2)

    P_xx = PHI + np.exp(-(x + 3.0)**2)
    P_yy = PHI + np.exp(-(x - 3.0)**2)
    P_xy = np.zeros(N)
    P_yx = np.zeros(N)

    print("=" * 90)
    print(f" SOLITON / COLLAPSE / UNKNOT — running to step {steps}")
    print(f" N={N}  dx={dx:.6f}  dt={dt:.6e}")
    print("=" * 90)
    print(f"{'step':>6} | {'max|Pxx|':>12} | {'FWHM':>10} | "
          f"{'max|I3|':>12} | {'sum I3':>14} | {'max|I2|':>12} | "
          f"{'max|I4|':>12}")
    print("-" * 90)

    width_history = []
    I3_history = []
    Pxx_history = []
    I2_history = []
    I4_history = []

    prev_Pxx_max = None
    prev_width = None

    for step in range(steps + 1):
        dPxx, dPxy, dPyx, dPyy = operator_v4(P_xx, P_xy, P_yx, P_yy, dx)

        P_xx = P_xx + dt * dPxx
        P_xy = P_xy + dt * dPxy
        P_yx = P_yx + dt * dPyx
        P_yy = P_yy + dt * dPyy

        if step % sample_every == 0 or step == steps:
            I2 = P_xx**2 + P_xy**2 + P_yx**2 + P_yy**2
            I3 = P_xx * P_yy - P_xy * P_yx
            I4 = P_xx**4 + P_yy**4

            mx_Pxx = float(np.max(np.abs(P_xx)))
            mx_I2  = float(np.max(np.abs(I2)))
            mx_I3  = float(np.max(np.abs(I3)))
            sum_I3 = float(np.sum(I3))
            mx_I4  = float(np.max(np.abs(I4)))

            w = fwhm(P_xx, x)

            width_history.append((step, w))
            I3_history.append((step, mx_I3, sum_I3))
            Pxx_history.append((step, mx_Pxx))
            I2_history.append((step, mx_I2))
            I4_history.append((step, mx_I4))

            w_str = f"{w:.6f}" if w is not None else "n/a"
            print(f"{step:>6} | {mx_Pxx:>12.4e} | {w_str:>10} | "
                  f"{mx_I3:>12.4e} | {sum_I3:>14.6e} | {mx_I2:>12.4e} | "
                  f"{mx_I4:>12.4e}")

            prev_Pxx_max = mx_Pxx
            prev_width = w

        finite = (np.isfinite(P_xx).all() and np.isfinite(P_yy).all()
                  and np.isfinite(P_xy).all() and np.isfinite(P_yx).all())
        if not finite:
            print(f"\nABORT: non-finite value at step {step}.")
            break

    return {
        "width_history": width_history,
        "I3_history":    I3_history,
        "Pxx_history":   Pxx_history,
        "I2_history":    I2_history,
        "I4_history":    I4_history,
        "final_step":    step,
    }


result = run(steps=20000, sample_every=1000)


# =============================================================================
# TEST 1 — WIDTH TREND
# =============================================================================
print()
print("=" * 90)
print(" TEST 1 — WIDTH TREND")
print("=" * 90)

widths = [(s, w) for s, w in result["width_history"] if w is not None]
if len(widths) >= 3:
    s0, w0 = widths[0]
    s1, w1 = widths[len(widths)//2]
    s2, w2 = widths[-1]

    print(f"  Initial width (step {s0:>5}): {w0:.6f}")
    print(f"  Middle  width (step {s1:>5}): {w1:.6f}")
    print(f"  Final   width (step {s2:>5}): {w2:.6f}")
    print()

    if w2 < w0 * 0.9:
        verdict = "WIDTH SHRINKING -> collapse signature"
    elif w2 > w0 * 1.1:
        verdict = "WIDTH GROWING -> dispersion signature"
    else:
        verdict = "WIDTH STABLE -> soliton signature"
    print(f"  {verdict}")
else:
    print("  Not enough width samples.")
print("=" * 90)


# =============================================================================
# TEST 2 — TOPOLOGICAL INVARIANT I3 TREND
# =============================================================================
print()
print("=" * 90)
print(" TEST 2 — TOPOLOGICAL INVARIANT I3 TREND")
print("=" * 90)

I3s = result["I3_history"]
if len(I3s) >= 3:
    s0, mx0, sum0 = I3s[0]
    s_last, mx_last, sum_last = I3s[-1]

    print(f"  max|I3|  at step {s0:>5}: {mx0:.6e}")
    print(f"  max|I3|  at step {s_last:>5}: {mx_last:.6e}")
    print(f"  sum I3   at step {s0:>5}: {sum0:+.6e}")
    print(f"  sum I3   at step {s_last:>5}: {sum_last:+.6e}")
    print()

    # monotonic check on max|I3|
    mx_vals = [m for _, m, _ in I3s]
    diffs = np.diff(mx_vals)
    monotonic_up = np.all(diffs > 0)
    monotonic_down = np.all(diffs < 0)
    oscillating = not monotonic_up and not monotonic_down

    if monotonic_up:
        verdict = "max|I3| MONOTONIC UP -> topology changing"
    elif monotonic_down:
        verdict = "max|I3| MONOTONIC DOWN -> topology changing (decreasing)"
    else:
        # count sign changes in diffs
        sign_changes = np.sum(np.diff(np.sign(diffs)) != 0)
        verdict = f"max|I3| NON-MONOTONIC ({sign_changes} sign changes) -> topology oscillating or preserved"
    print(f"  {verdict}")

    # sum I3 sign flip check
    sum_vals = [sm for _, _, sm in I3s]
    if any(sm > 0 for sm in sum_vals) and any(sm < 0 for sm in sum_vals):
        print(f"  sum I3 SIGN FLIPS -> topology crossing zero")
    else:
        print(f"  sum I3 SIGN STABLE -> topology not flipping")
print("=" * 90)


# =============================================================================
# TEST 3 — RELEASE OR CONTINUED GROWTH
# =============================================================================
print()
print("=" * 90)
print(" TEST 3 — RELEASE OR CONTINUED GROWTH")
print("=" * 90)

Pxxs = result["Pxx_history"]
if len(Pxxs) >= 3:
    vals = [v for _, v in Pxxs]
    diffs = np.diff(vals)

    # find any negative diff (a drop)
    drops = [(Pxxs[i+1][0], diffs[i]) for i in range(len(diffs)) if diffs[i] < 0]
    if drops:
        print(f"  DROPS detected ({len(drops)} samples with negative growth):")
        for step, d in drops[:5]:
            print(f"    step {step}: d(max|Pxx|) = {d:+.6e}")
        print(f"  RELEASE SIGNATURE -> the field decreased at least once.")
        print(f"  This is consistent with unknotting.")
    else:
        print(f"  No drops. max|Pxx| grew monotonically across all samples.")
        print(f"  CONTINUED GROWTH -> heading toward collapse.")

    # growth rate at the end
    if len(vals) >= 3:
        late_rate = (vals[-1] - vals[-2]) / vals[-2]
        early_rate = (vals[1] - vals[0]) / vals[0] if vals[0] > 0 else 0
        print()
        print(f"  early growth rate (first 2 samples) : {early_rate*100:+.3f}%")
        print(f"  late  growth rate (last 2 samples)  : {late_rate*100:+.3f}%")
        if late_rate > early_rate * 1.5:
            print(f"  ACCELERATING -> heading toward collapse")
        elif late_rate < early_rate * 0.5:
            print(f"  DECELERATING -> approaching plateau")
        else:
            print(f"  STEADY -> linear growth regime")
print("=" * 90)
print()
print(f" Final step reached: {result['final_step']}")
print("=" * 90)
```

```python
"""
ENVELOPE OPTIONS TEST — THREE VARIANTS
========================================

Option A: pre-clip the invariants (mechanical fix)
    I2_safe = clip(I2, None, 1e150)
    I4_safe = clip(I4, None, 1e75)
    exp_arg = -0.5 * (I2_safe**2 + I4_safe**4)

Option B: soften the envelope power (theoretical change)
    exp_arg = -0.5 * (I2**2 + I4**2)

Option C: no fix. Record the abort as the pivot marker.
    exp_arg = -0.5 * (I2**2 + I4**4)  (may overflow)

Same operator otherwise. Same diagnostics.
"""

import numpy as np


PHI       = 1.618033988749895
PI_MAX    = 5.9259
KO_SIGMA  = 0.045

ALPHA_HYB = 1.0
BETA_HYB  = 0.1
GAMMA_HYB = 0.1
I_G       = 1.0

IK_COEFF  = 1.0 + 1.01
I2_PIVOT  = PI_MAX
IK_PIVOT  = PI_MAX


def dPhi_hyb_dP_yx(P_yx, I1):
    g = I1**2 / (I1**2 + I_G**2)
    abs_p  = np.abs(P_yx)
    sign_p = np.sign(P_yx)
    num = 2.0 * P_yx * (1.0 + GAMMA_HYB * abs_p) - GAMMA_HYB * sign_p * P_yx**2
    den = (1.0 + GAMMA_HYB * abs_p) ** 2
    return ALPHA_HYB + g * BETA_HYB * num / den


def _KO(f, dx):
    stencil = (np.roll(f, -2) - 4.0*np.roll(f, -1) + 6.0*f
               - 4.0*np.roll(f, 1) + np.roll(f, 2))
    return KO_SIGMA * stencil / (dx**4)


# =============================================================================
# ENVELOPE VARIANTS
# =============================================================================
def envelope_A(I2, I4):
    """Option A: pre-clip invariants before the power."""
    I2_safe = np.clip(I2, None, 1e150)
    I4_safe = np.clip(I4, None, 1e75)
    arg = -0.5 * (I2_safe**2 + I4_safe**4)
    return np.exp(np.clip(arg, -50.0, 0.0))


def envelope_B(I2, I4):
    """Option B: soften the envelope power. I4^2 instead of I4^4."""
    arg = -0.5 * (I2**2 + I4**2)
    return np.exp(np.clip(arg, -50.0, 0.0))


def envelope_C(I2, I4):
    """Option C: no fix. May overflow."""
    arg = -0.5 * (I2**2 + I4**4)
    return np.exp(np.clip(arg, -50.0, 0.0))


# =============================================================================
# OPERATOR FACTORY
# =============================================================================
def make_operator(envelope_fn, label):
    def operator(P_xx, P_xy, P_yx, P_yy, dx):
        I1 = P_xx + P_yy
        I2 = P_xx**2 + P_xy**2 + P_yx**2 + P_yy**2
        I4 = P_xx**4 + P_yy**4
        I1_safe = np.clip(I1, 1e-3, None)

        def D(f):
            return (np.roll(f, -1) - np.roll(f, 1)) / (2.0 * dx)

        with np.errstate(over='ignore', invalid='ignore'):
            envelope = envelope_fn(I2, I4)

        sat2 = 1.0 + np.maximum(I2, 0.0) / I2_PIVOT

        def C_term(P_comp, D_P):
            I_k = IK_COEFF * P_comp
            I_k_sq = I_k ** 2
            sat3 = 1.0 + I_k_sq / (IK_PIVOT ** 2)
            term1 = 0.2 * (D_P * I1)
            term2 = 0.2 * (I2 - I1) * (I1 + I2) / sat2
            term3 = 0.1 * I_k_sq / sat3
            term4 = ((1.0 / PI_MAX) * (I1_safe**(-0.5) - 1.0)
                     * envelope * P_comp)
            return term1 + term2 + term3 + term4

        dPxx = C_term(P_xx, D(P_xx))
        dPxy = C_term(P_xy, D(P_xy))
        dPyx = C_term(P_yx, D(P_yx))
        dPyy = C_term(P_yy, D(P_yy))
        dPyx = dPyx + dPhi_hyb_dP_yx(P_yx, I1)

        return (dPxx - _KO(P_xx, dx),
                dPxy - _KO(P_xy, dx),
                dPyx - _KO(P_yx, dx),
                dPyy - _KO(P_yy, dx))
    operator.__label__ = label
    return operator


# =============================================================================
# FWHM DIAGNOSTIC
# =============================================================================
def fwhm(profile, x):
    prof = np.abs(profile)
    max_idx = int(np.argmax(prof))
    max_val = prof[max_idx]
    if max_val <= 0:
        return None
    half = 0.5 * max_val
    left = max_idx
    while left > 0 and prof[left] > half:
        left -= 1
    right = max_idx
    while right < len(prof) - 1 and prof[right] > half:
        right += 1
    return x[right] - x[left]


# =============================================================================
# RUN
# =============================================================================
def run(operator, label, steps=50000, N=256, L=20.0,
        dt_coeff=0.005, sample_every=2500):
    x  = np.linspace(-L/2, L/2, N, endpoint=False)
    dx = L / N
    dt = dt_coeff * (dx ** 2)

    P_xx = PHI + np.exp(-(x + 3.0)**2)
    P_yy = PHI + np.exp(-(x - 3.0)**2)
    P_xy = np.zeros(N)
    P_yx = np.zeros(N)

    print("=" * 100)
    print(f" {label}   (max steps = {steps})")
    print("=" * 100)
    print(f"{'step':>7} | {'max|Pxx|':>12} | {'FWHM':>10} | "
          f"{'max|I2|':>12} | {'max|I4|':>12} | {'sum I3':>14}")
    print("-" * 100)

    prev_Pxx = None
    aborted_at = None
    drop_count = 0

    for step in range(steps + 1):
        try:
            dPxx, dPxy, dPyx, dPyy = operator(P_xx, P_xy, P_yx, P_yy, dx)
        except Exception as e:
            print(f"  Operator exception at step {step}: {e}")
            aborted_at = step
            break

        with np.errstate(over='ignore', invalid='ignore'):
            P_xx = P_xx + dt * dPxx
            P_xy = P_xy + dt * dPxy
            P_yx = P_yx + dt * dPyx
            P_yy = P_yy + dt * dPyy

        if step % sample_every == 0 or step == steps:
            I2 = P_xx**2 + P_xy**2 + P_yx**2 + P_yy**2
            I3 = P_xx * P_yy - P_xy * P_yx
            I4 = P_xx**4 + P_yy**4

            mx_Pxx = float(np.max(np.abs(P_xx)))
            mx_I2  = float(np.max(np.abs(I2)))
            mx_I4  = float(np.max(np.abs(I4)))
            sum_I3 = float(np.sum(I3))

            w = fwhm(P_xx, x)
            w_str = f"{w:.6f}" if w is not None else "n/a"

            dP = (mx_Pxx - prev_Pxx) if prev_Pxx is not None else 0.0
            if dP < 0:
                drop_count += 1

            print(f"{step:>7} | {mx_Pxx:>12.4e} | {w_str:>10} | "
                  f"{mx_I2:>12.4e} | {mx_I4:>12.4e} | {sum_I3:>14.6e}")

            prev_Pxx = mx_Pxx

        finite = (np.isfinite(P_xx).all() and np.isfinite(P_yy).all()
                  and np.isfinite(P_xy).all() and np.isfinite(P_yx).all())
        if not finite:
            print(f"  ABORT: non-finite value at step {step}.")
            aborted_at = step
            break

    print("-" * 100)
    print(f"  Final step reached  : {step if aborted_at is None else aborted_at}")
    print(f"  Aborted             : {aborted_at is not None}")
    print(f"  Drops detected      : {drop_count}")
    print("=" * 100)
    return {"label": label, "aborted_at": aborted_at,
            "final_step": step if aborted_at is None else aborted_at,
            "drop_count": drop_count}


# =============================================================================
# RUN ALL THREE
# =============================================================================
op_A = make_operator(envelope_A, "Option A — pre-clip invariants")
op_B = make_operator(envelope_B, "Option B — softened envelope (I4^2)")
op_C = make_operator(envelope_C, "Option C — no fix")

result_A = run(op_A, op_A.__label__)
print()
result_B = run(op_B, op_B.__label__)
print()
result_C = run(op_C, op_C.__label__)


print()
print("#" * 100)
print("# COMPARISON")
print("#" * 100)
for r in [result_A, result_B, result_C]:
    print(f"  {r['label']:45s}  aborted_at = {r['aborted_at']}  "
          f"final_step = {r['final_step']}  drops = {r['drop_count']}")
print("#" * 100)
```

```python
"""
CORRECTED ENGINE v5 — PRE-CLIP + ADAPTIVE dt (spatial gradient)
================================================================

Fixes:
  FIX 1: pre-clip P_xx, P_yy, P_xy, P_yx at 1e19 before raising to powers.
         Prevents P**4 and I4**4 from overflowing float64.

  FIX 2: adaptive dt tied to the spatial gradient of P_xx.
         dt = CFL_FACTOR * dx / (1 + max|D(P_xx)| / PI_MAX)
         CFL-appropriate. Shrinks as the peak sharpens.

Everything else aligned with v4:
  - Envelope exp(-0.5 * (I2**2 + I4**4)) clipped to [-50, 0]
  - Saturations sat2 (on I2) and sat3 (on I_k^2)
  - KO dissipation, dissipative sign
  - Torque channel dPyx += dPhi_hyb_dP_yx
  - I_k = 2.01 * P_comp
"""

import os
import shutil
import json
import numpy as np
from datetime import datetime


try:
    from google.colab import drive, files
    IN_COLAB = True
    drive.mount('/content/drive')
except ImportError:
    IN_COLAB = False
    print("Not in Colab. Drive and download steps will be simulated.")

PROJECT_NAME = "FRCMD_FROZEN_CORE_V5"
timestamp    = datetime.now().strftime("%Y%m%d_%H%M%S")
output_dir   = f"output_{timestamp}"
os.makedirs(output_dir, exist_ok=True)
print(f"Workspace: {output_dir}/")


# =============================================================================
# CONSTANTS
# =============================================================================
PHI       = 1.618033988749895
PI_MAX    = 5.9259
KO_SIGMA  = 0.045

ALPHA_HYB = 1.0
BETA_HYB  = 0.1
GAMMA_HYB = 0.1
I_G       = 1.0

IK_COEFF  = 1.0 + 1.01        # = 2.01

I2_PIVOT  = PI_MAX
IK_PIVOT  = PI_MAX

FIELD_CLIP = 1.0e19           # [CHOSEN] so that P**16 < 1e308

CFL_FACTOR = 0.4              # [CHOSEN]
DT_MIN     = 1.0e-12          # [CHOSEN]
N_DEFAULT  = 256
L_DEFAULT  = 20.0
DT_MAX     = 0.005 * (L_DEFAULT / N_DEFAULT) ** 2


# =============================================================================
# OPERATORS
# =============================================================================
def dPhi_hyb_dP_yx(P_yx, I1):
    g = I1**2 / (I1**2 + I_G**2)
    abs_p  = np.abs(P_yx)
    sign_p = np.sign(P_yx)
    num = 2.0 * P_yx * (1.0 + GAMMA_HYB * abs_p) - GAMMA_HYB * sign_p * P_yx**2
    den = (1.0 + GAMMA_HYB * abs_p) ** 2
    return ALPHA_HYB + g * BETA_HYB * num / den


def _KO(f, dx):
    stencil = (np.roll(f, -2) - 4.0*np.roll(f, -1) + 6.0*f
               - 4.0*np.roll(f, 1) + np.roll(f, 2))
    return KO_SIGMA * stencil / (dx**4)


def operator_v5(P_xx, P_xy, P_yx, P_yy, dx):
    # FIX 1: pre-clip before powers
    P_xx_safe = np.clip(P_xx, -FIELD_CLIP, FIELD_CLIP)
    P_yy_safe = np.clip(P_yy, -FIELD_CLIP, FIELD_CLIP)
    P_xy_safe = np.clip(P_xy, -FIELD_CLIP, FIELD_CLIP)
    P_yx_safe = np.clip(P_yx, -FIELD_CLIP, FIELD_CLIP)

    I1 = P_xx_safe + P_yy_safe
    I2 = (P_xx_safe**2 + P_xy_safe**2
          + P_yx_safe**2 + P_yy_safe**2)
    I4 = P_xx_safe**4 + P_yy_safe**4
    I1_safe = np.clip(I1, 1e-3, None)

    def D(f):
        return (np.roll(f, -1) - np.roll(f, 1)) / (2.0 * dx)

    exp_arg = -0.5 * (I2**2 + I4**4)
    exp_arg = np.clip(exp_arg, -50.0, 0.0)
    envelope = np.exp(exp_arg)

    sat2 = 1.0 + np.maximum(I2, 0.0) / I2_PIVOT

    def C_term(P_comp_safe, P_comp_raw, D_P):
        I_k = IK_COEFF * P_comp_safe
        I_k_sq = I_k ** 2
        sat3 = 1.0 + I_k_sq / (IK_PIVOT ** 2)
        term1 = 0.2 * (D_P * I1)
        term2 = 0.2 * (I2 - I1) * (I1 + I2) / sat2
        term3 = 0.1 * I_k_sq / sat3
        term4 = ((1.0 / PI_MAX) * (I1_safe**(-0.5) - 1.0)
                 * envelope * P_comp_raw)
        return term1 + term2 + term3 + term4

    dPxx = C_term(P_xx_safe, P_xx, D(P_xx_safe))
    dPxy = C_term(P_xy_safe, P_xy, D(P_xy_safe))
    dPyx = C_term(P_yx_safe, P_yx, D(P_yx_safe))
    dPyy = C_term(P_yy_safe, P_yy, D(P_yy_safe))
    dPyx = dPyx + dPhi_hyb_dP_yx(P_yx_safe, I1)

    return (dPxx - _KO(P_xx_safe, dx),
            dPxy - _KO(P_xy_safe, dx),
            dPyx - _KO(P_yx_safe, dx),
            dPyy - _KO(P_yy_safe, dx))


# =============================================================================
# FWHM
# =============================================================================
def fwhm(profile, x):
    prof = np.abs(profile)
    max_idx = int(np.argmax(prof))
    max_val = prof[max_idx]
    if max_val <= 0:
        return None
    half = 0.5 * max_val
    left = max_idx
    while left > 0 and prof[left] > half:
        left -= 1
    right = max_idx
    while right < len(prof) - 1 and prof[right] > half:
        right += 1
    return x[right] - x[left]


# =============================================================================
# INTEGRATION
# =============================================================================
def run(steps=100000, N=N_DEFAULT, L=L_DEFAULT, sample_every=2500):
    x  = np.linspace(-L/2, L/2, N, endpoint=False)
    dx = L / N

    P_xx = PHI + np.exp(-(x + 3.0)**2)
    P_yy = PHI + np.exp(-(x - 3.0)**2)
    P_xy = np.zeros(N)
    P_yx = np.zeros(N)

    dt = DT_MAX
    aborted_at = None
    drops = 0
    prev_Pxx = None
    dt_history = []

    print("=" * 100)
    print(f" CORRECTED ENGINE v5 — pre-clip {FIELD_CLIP:.0e} + adaptive dt (spatial grad)")
    print(f" N={N}  dx={dx:.6f}  dt ceiling={DT_MAX:.6e}  max steps={steps}")
    print("=" * 100)
    print(f"{'step':>7} | {'dt':>10} | {'max|Pxx|':>11} | "
          f"{'FWHM':>10} | {'max|I2|':>11} | {'max|I4|':>11} | "
          f"{'max|D_Pxx|':>12}")
    print("-" * 100)

    for step in range(steps + 1):
        dPxx, dPxy, dPyx, dPyy = operator_v5(P_xx, P_xy, P_yx, P_yy, dx)

        # FIX 2: adaptive dt using the spatial gradient of P_xx
        D_Pxx = (np.roll(P_xx, -1) - np.roll(P_xx, 1)) / (2.0 * dx)
        max_grad = float(np.max(np.abs(D_Pxx)))

        dt_new = CFL_FACTOR * dx / (1.0 + max_grad / PI_MAX)
        dt_new = max(DT_MIN, min(DT_MAX, dt_new))
        dt = dt_new

        P_xx = P_xx + dt * dPxx
        P_xy = P_xy + dt * dPxy
        P_yx = P_yx + dt * dPyx
        P_yy = P_yy + dt * dPyy

        if step % sample_every == 0 or step == steps:
            I2_now = P_xx**2 + P_xy**2 + P_yx**2 + P_yy**2
            I4_now = P_xx**4 + P_yy**4
            mx_Pxx = float(np.max(np.abs(P_xx)))
            mx_I2  = float(np.max(np.abs(I2_now)))
            mx_I4  = float(np.max(np.abs(I4_now)))

            w = fwhm(P_xx, x)
            w_str = f"{w:.6f}" if w is not None else "n/a"

            if prev_Pxx is not None and mx_Pxx < prev_Pxx:
                drops += 1
            prev_Pxx = mx_Pxx

            dt_history.append((step, dt))

            print(f"{step:>7} | {dt:>10.3e} | {mx_Pxx:>11.4e} | "
                  f"{w_str:>10} | {mx_I2:>11.4e} | {mx_I4:>11.4e} | "
                  f"{max_grad:>12.4e}")

        finite = (np.isfinite(P_xx).all() and np.isfinite(P_yy).all()
                  and np.isfinite(P_xy).all() and np.isfinite(P_yx).all())
        if not finite:
            print(f"  ABORT: non-finite value at step {step}.")
            aborted_at = step
            break

    print("-" * 100)
    final_step = step if aborted_at is None else aborted_at
    print(f"  Final step reached  : {final_step}")
    print(f"  Aborted             : {aborted_at is not None}")
    print(f"  Drops detected      : {drops}")
    if dt_history:
        print(f"  dt at start         : {dt_history[0][1]:.6e}")
        print(f"  dt at end           : {dt_history[-1][1]:.6e}")
        print(f"  dt ratio            : {dt_history[-1][1] / dt_history[0][1]:.6e}")
    print("=" * 100)

    return {
        "final_step":  final_step,
        "aborted_at":  aborted_at,
        "drops":       drops,
        "dt_history":  dt_history,
    }


# =============================================================================
# RUN
# =============================================================================
result = run(steps=100000, sample_every=2500)


# =============================================================================
# OBSERVATION RECORD
# =============================================================================
obs_path = os.path.join(output_dir, "observation.json")
with open(obs_path, "w") as f:
    json.dump({
        "engine":         "frozen_core_v5",
        "final_step":     int(result["final_step"]),
        "aborted":        bool(result["aborted_at"] is not None),
        "aborted_at":     int(result["aborted_at"]) if result["aborted_at"] else None,
        "drops":          int(result["drops"]),
        "FIELD_CLIP":     float(FIELD_CLIP),
        "CFL_FACTOR":     float(CFL_FACTOR),
        "DT_MIN":         float(DT_MIN),
        "DT_MAX":         float(DT_MAX),
        "dt_start":       float(result["dt_history"][0][1]) if result["dt_history"] else None,
        "dt_end":         float(result["dt_history"][-1][1]) if result["dt_history"] else None,
        "status":         "OBSERVATION_ONLY",
        "timestamp":      datetime.now().isoformat(),
    }, f, indent=2)


# =============================================================================
# PRESERVATION
# =============================================================================
zip_base = f"{PROJECT_NAME}_{timestamp}"
zip_path = shutil.make_archive(zip_base, 'zip', output_dir)

drive_ok = False
if IN_COLAB:
    try:
        drive_dir = f"/content/drive/MyDrive/{PROJECT_NAME}/{output_dir}"
        os.makedirs(drive_dir, exist_ok=True)
        for item in os.listdir(output_dir):
            shutil.copy2(os.path.join(output_dir, item),
                         os.path.join(drive_dir, item))
        shutil.copy2(zip_path, f"/content/drive/MyDrive/{PROJECT_NAME}/{zip_base}.zip")
        drive_ok = True
    except Exception as e:
        print(f"Drive backup failed: {e}")

download_ok = False
if IN_COLAB:
    try:
        files.download(zip_path)
        download_ok = True
    except Exception as e:
        print(f"Download failed: {e}")


# =============================================================================
# BENCH READOUT
# =============================================================================
print()
print("=" * 70)
print(" FRCMD FROZEN CORE v5 — BENCH READOUT")
print("=" * 70)
print(f" Output directory : {os.path.abspath(output_dir)}")
print(f" ZIP archive      : {os.path.abspath(zip_path)}")
print(f" Drive backup     : {drive_ok}")
print(f" Download         : {download_ok}")
print(f" Final step       : {result['final_step']}")
print(f" Aborted          : {result['aborted_at'] is not None}")
print(f" Drops            : {result['drops']}")
print(f" FIELD_CLIP       : {FIELD_CLIP:.0e}")
print(f" CFL_FACTOR       : {CFL_FACTOR}")
print(f" DT_MIN           : {DT_MIN:.0e}")
print(f" DT_MAX           : {DT_MAX:.6e}")
if result["dt_history"]:
    print(f" dt start         : {result['dt_history'][0][1]:.6e}")
    print(f" dt end           : {result['dt_history'][-1][1]:.6e}")
print("=" * 70)
```

```python
"""
CELL 3 v7 — HORIZON DERIVATION TEST HARNESS
=============================================

Purpose:
  Sweep the initial amplitude A across a range. For each A, run the v6
  engine to a fixed number of steps. Log M_Pi, R_s_target, R_peak,
  min_C_sq, max_grad at each sample step.

  The success condition: is R_peak(A) proportional to M_Pi(A)?
  If yes, R_peak = (2 G_eff / c_eff^2) * M_Pi, and the horizon is a
  native pivot of the Pi-field.

Calibration constants (PROVISIONAL-NOTATION / HYPOTHESIS):
  S_X = 1.0e-35 m    (1 lattice unit of dx = S_X meters)
  S_M = 1.0e-8 kg    (1 unit of integrated |P_xx| dx = S_M kilograms)

  These are NOT sourced from the record. They are choices that set the
  scale. If the derivation succeeds with these values, the framework has
  a calibrated mass functional. If it fails, the values or the functional
  need adjustment.
"""

import os
import shutil
import json
import numpy as np
from datetime import datetime


try:
    from google.colab import drive, files
    IN_COLAB = True
    drive.mount('/content/drive')
except ImportError:
    IN_COLAB = False
    print("Not in Colab. Drive and download steps will be simulated.")


# =============================================================================
# PROJECT
# =============================================================================
PROJECT_NAME = "FRCMD_HORIZON_HARNESS_V7"
timestamp    = datetime.now().strftime("%Y%m%d_%H%M%S")
output_dir   = f"output_{timestamp}"
os.makedirs(output_dir, exist_ok=True)
print(f"Workspace: {output_dir}/")


# =============================================================================
# CALIBRATION (PROVISIONAL-NOTATION / HYPOTHESIS)
# =============================================================================
S_X = 1.0e-35      # m per lattice unit
S_M = 1.0e-8       # kg per unit of integrated |P_xx| dx
G_CLASSICAL = 6.67430e-11
C_CLASSICAL = 299792458.0


# =============================================================================
# FROZEN CORE CONSTANTS
# =============================================================================
PHI       = 1.618033988749895
PI_MAX    = 5.9259
KO_SIGMA  = 0.045

ALPHA_HYB = 1.0
BETA_HYB  = 0.1
GAMMA_HYB = 0.1
I_G       = 1.0

IK_COEFF  = 1.0 + 1.01
I2_PIVOT  = PI_MAX
IK_PIVOT  = PI_MAX

FIELD_CLIP = 1.0e19

CFL_FACTOR = 0.4
DT_MIN     = 1.0e-12
N_DEFAULT  = 256
L_DEFAULT  = 20.0
DT_MAX     = 0.005 * (L_DEFAULT / N_DEFAULT) ** 2


# =============================================================================
# OPERATORS
# =============================================================================
def dPhi_hyb_dP_yx(P_yx, I1):
    g = I1**2 / (I1**2 + I_G**2)
    abs_p  = np.abs(P_yx)
    sign_p = np.sign(P_yx)
    num = 2.0 * P_yx * (1.0 + GAMMA_HYB * abs_p) - GAMMA_HYB * sign_p * P_yx**2
    den = (1.0 + GAMMA_HYB * abs_p) ** 2
    return ALPHA_HYB + g * BETA_HYB * num / den


def _KO(f, dx):
    stencil = (np.roll(f, -2) - 4.0*np.roll(f, -1) + 6.0*f
               - 4.0*np.roll(f, 1) + np.roll(f, 2))
    return KO_SIGMA * stencil / (dx**4)


def operator_v6(P_xx, P_xy, P_yx, P_yy, dx):
    P_xx_safe = np.clip(P_xx, -FIELD_CLIP, FIELD_CLIP)
    P_yy_safe = np.clip(P_yy, -FIELD_CLIP, FIELD_CLIP)
    P_xy_safe = np.clip(P_xy, -FIELD_CLIP, FIELD_CLIP)
    P_yx_safe = np.clip(P_yx, -FIELD_CLIP, FIELD_CLIP)

    I1 = P_xx_safe + P_yy_safe
    I2 = (P_xx_safe**2 + P_xy_safe**2
          + P_yx_safe**2 + P_yy_safe**2)
    I4 = P_xx_safe**4 + P_yy_safe**4
    I1_safe = np.clip(I1, 1e-3, None)

    def D(f):
        return (np.roll(f, -1) - np.roll(f, 1)) / (2.0 * dx)

    exp_arg = -0.5 * (I2**2 + I4**4)
    exp_arg = np.clip(exp_arg, -50.0, 0.0)
    envelope = np.exp(exp_arg)

    sat2 = 1.0 + (I2 ** 2) / (I2_PIVOT ** 2)

    def C_term(P_comp_safe, P_comp_raw, D_P):
        I_k = IK_COEFF * P_comp_safe
        I_k_sq = I_k ** 2
        sat3 = 1.0 + I_k_sq / (IK_PIVOT ** 2)
        term1 = 0.2 * (D_P * I1)
        term2 = 0.2 * (I2 - I1) * (I1 + I2) / sat2
        term3 = 0.1 * I_k_sq / sat3
        term4 = ((1.0 / PI_MAX) * (I1_safe**(-0.5) - 1.0)
                 * envelope * P_comp_raw)
        return term1 + term2 + term3 + term4

    dPxx = C_term(P_xx_safe, P_xx, D(P_xx_safe))
    dPxy = C_term(P_xy_safe, P_xy, D(P_xy_safe))
    dPyx = C_term(P_yx_safe, P_yx, D(P_yx_safe))
    dPyy = C_term(P_yy_safe, P_yy, D(P_yy_safe))
    dPyx = dPyx + dPhi_hyb_dP_yx(P_yx_safe, I1)

    return (dPxx - _KO(P_xx_safe, dx),
            dPxy - _KO(P_xy_safe, dx),
            dPyx - _KO(P_yx_safe, dx),
            dPyy - _KO(P_yy_safe, dx))


# =============================================================================
# SINGLE RUN
# =============================================================================
def run_single(A, steps=20000, N=N_DEFAULT, L=L_DEFAULT,
               sample_every=1000, verbose=False):
    """
    Run the v6 engine with initial amplitude A.
    Returns the horizon log for this run.
    """
    x  = np.linspace(-L/2, L/2, N, endpoint=False)
    dx = L / N

    P_xx = PHI + A * np.exp(-(x + 3.0)**2)
    P_yy = PHI + A * np.exp(-(x - 3.0)**2)
    P_xy = np.zeros(N)
    P_yx = np.zeros(N)

    dt = DT_MAX
    log = []

    for step in range(steps + 1):
        dPxx, dPxy, dPyx, dPyy = operator_v6(P_xx, P_xy, P_yx, P_yy, dx)

        D_Pxx = (np.roll(P_xx, -1) - np.roll(P_xx, 1)) / (2.0 * dx)
        max_grad = float(np.max(np.abs(D_Pxx)))

        dt_new = CFL_FACTOR * dx / (1.0 + max_grad / PI_MAX)
        dt_new = max(DT_MIN, min(DT_MAX, dt_new))
        dt = dt_new

        P_xx = P_xx + dt * dPxx
        P_xy = P_xy + dt * dPxy
        P_yx = P_yx + dt * dPyx
        P_yy = P_yy + dt * dPyy

        if step % sample_every == 0 or step == steps:
            # DERIVED HORIZON FUNCTIONAL & SCALING
            rho_pi = np.abs(P_xx) + np.abs(P_yy)
            M_Pi = S_M * np.sum(rho_pi) * (dx * S_X)
            R_s_target = 2.0 * G_CLASSICAL * M_Pi / (C_CLASSICAL ** 2)

            r_phys = np.abs(x) * S_X
            r_safe = np.maximum(r_phys, 1e-40)
            prefactor = 2.0 * G_CLASSICAL * M_Pi / (C_CLASSICAL ** 2)
            C_sq_effective = 1.0 - (prefactor / r_safe)
            min_C_sq = float(np.min(C_sq_effective))

            peak_idx = int(np.argmax(np.abs(P_xx)))
            R_peak = float(np.abs(x[peak_idx]) * S_X)

            log.append({
                "step":         step,
                "A":            float(A),
                "M_Pi_kg":      float(M_Pi),
                "R_s_target_m": float(R_s_target),
                "R_peak_m":     float(R_peak),
                "min_C_sq":     float(min_C_sq),
                "max_grad":     float(max_grad),
                "max_Pxx":      float(np.max(np.abs(P_xx))),
            })

        finite = (np.isfinite(P_xx).all() and np.isfinite(P_yy).all()
                  and np.isfinite(P_xy).all() and np.isfinite(P_yx).all())
        if not finite:
            if verbose:
                print(f"  A={A}  ABORT at step {step}")
            break

    if verbose:
        print(f"  A={A:.3f}  final_step={step}  "
              f"M_Pi={log[-1]['M_Pi_kg']:.4e}  "
              f"R_s={log[-1]['R_s_target_m']:.4e}  "
              f"R_peak={log[-1]['R_peak_m']:.4e}  "
              f"minC^2={log[-1]['min_C_sq']:.6f}")

    return log


# =============================================================================
# PARAMETER SWEEP
# =============================================================================
def run_sweep(amplitudes, steps=20000, sample_every=1000):
    """
    Run a sweep of initial amplitudes A.
    Returns the concatenated horizon log.
    """
    print("=" * 100)
    print(f" HORIZON DERIVATION SWEEP — {len(amplitudes)} amplitudes")
    print(f" S_X = {S_X:.0e} m/lattice-unit   S_M = {S_M:.0e} kg/unit-load")
    print(f" steps = {steps}  sample_every = {sample_every}")
    print("=" * 100)
    print(f"{'A':>8} | {'final_M_Pi':>13} | {'R_s_target':>13} | "
          f"{'R_peak':>13} | {'min_C^2':>11} | {'ratio':>10}")
    print("-" * 100)

    full_log = []
    summary = []

    for A in amplitudes:
        log = run_single(A, steps=steps, sample_every=sample_every)
        if log:
            last = log[-1]
            ratio = (last["R_peak_m"] / last["R_s_target_m"]
                     if last["R_s_target_m"] > 0 else float("nan"))
            print(f"{A:>8.3f} | {last['M_Pi_kg']:>13.4e} | "
                  f"{last['R_s_target_m']:>13.4e} | "
                  f"{last['R_peak_m']:>13.4e} | "
                  f"{last['min_C_sq']:>11.6f} | {ratio:>10.4e}")
            full_log.extend(log)
            summary.append(last)

    print("-" * 100)
    return full_log, summary


# =============================================================================
# RUN
# =============================================================================
# Sweep of initial amplitudes
AMPLITUDES = [0.1, 0.5, 1.0, 2.0, 5.0, 10.0, 20.0, 50.0]
STEPS = 20000
SAMPLE_EVERY = 1000

horizon_log, summary = run_sweep(AMPLITUDES, steps=STEPS,
                                 sample_every=SAMPLE_EVERY)


# =============================================================================
# ANALYSIS — is R_peak proportional to M_Pi?
# =============================================================================
print()
print("=" * 100)
print(" DERIVATION CHECK — is R_peak proportional to M_Pi?")
print("=" * 100)

if len(summary) >= 3:
    M_vals = np.array([s["M_Pi_kg"] for s in summary])
    R_vals = np.array([s["R_peak_m"] for s in summary])

    # linear fit through the origin: R = alpha * M
    # alpha = sum(M*R) / sum(M^2)
    denom = np.sum(M_vals ** 2)
    if denom > 0:
        alpha_fit = float(np.sum(M_vals * R_vals) / denom)
        # residuals
        R_fit = alpha_fit * M_vals
        residuals = R_vals - R_fit
        rel_resid = residuals / np.maximum(np.abs(R_vals), 1e-30)
        max_rel_resid = float(np.max(np.abs(rel_resid)))
        mean_rel_resid = float(np.mean(np.abs(rel_resid)))

        # expected alpha = 2 G / c^2
        alpha_Schw = 2.0 * G_CLASSICAL / (C_CLASSICAL ** 2)

        print(f"  Fitted alpha (R_peak = alpha * M_Pi) : {alpha_fit:.6e}")
        print(f"  Schwarzschild alpha (2G/c^2)         : {alpha_Schw:.6e}")
        print(f"  Ratio fitted / Schwarzschild         : {alpha_fit / alpha_Schw:.6e}")
        print()
        print(f"  Max relative residual  : {max_rel_resid:.6e}")
        print(f"  Mean relative residual : {mean_rel_resid:.6e}")
        print()
        if mean_rel_resid < 1e-2:
            print("  VERDICT: R_peak is proportional to M_Pi to <1%.")
        elif mean_rel_resid < 1e-1:
            print("  VERDICT: R_peak is proportional to M_Pi to <10%.")
        else:
            print("  VERDICT: R_peak is NOT simply proportional to M_Pi.")
            print("           The derivation as tested does not close.")
else:
    print("  Not enough summary points for a fit.")
print("=" * 100)


# =============================================================================
# WRITE LOGS
# =============================================================================
horizon_log_path = os.path.join(output_dir, "horizon_log.json")
with open(horizon_log_path, "w") as f:
    json.dump(horizon_log, f, indent=2)
print(f"Wrote {horizon_log_path}  ({len(horizon_log)} entries)")

summary_path = os.path.join(output_dir, "sweep_summary.json")
with open(summary_path, "w") as f:
    json.dump(summary, f, indent=2)
print(f"Wrote {summary_path}  ({len(summary)} entries)")


# =============================================================================
# PRESERVATION
# =============================================================================
zip_base = f"{PROJECT_NAME}_{timestamp}"
zip_path = shutil.make_archive(zip_base, 'zip', output_dir)

drive_ok = False
drive_dir = f"/content/drive/MyDrive/{PROJECT_NAME}/{output_dir}"
if IN_COLAB:
    try:
        os.makedirs(drive_dir, exist_ok=True)
        for item in os.listdir(output_dir):
            shutil.copy2(os.path.join(output_dir, item),
                         os.path.join(drive_dir, item))
        shutil.copy2(zip_path, f"/content/drive/MyDrive/{PROJECT_NAME}/{zip_base}.zip")
        drive_ok = True
    except Exception as e:
        print(f"Drive backup failed: {e}")

download_ok = False
if IN_COLAB:
    try:
        files.download(zip_path)
        download_ok = True
    except Exception as e:
        print(f"Download failed: {e}")


print()
print(f"{'✓' if os.path.exists(output_dir) else '✗'} Colab workspace saved")
print(f"{'✓' if drive_ok else '✗'} Google Drive backup saved")
print(f"{'✓' if download_ok else '✗'} Download package created")
print()
print("=" * 70)
print(" HORIZON DERIVATION HARNESS — BENCH READOUT")
print("=" * 70)
print(f" OUTPUT DIRECTORY: {os.path.abspath(output_dir)}")
print(f" GOOGLE DRIVE BACKUP: {drive_dir if drive_ok else 'FAILED'}")
print(f" MASTER ZIP: {os.path.abspath(zip_path)}")
print(f" AMPLITUDES: {AMPLITUDES}")
print(f" STEPS: {STEPS}")
print(f" STATUS: {'SUCCESS' if (drive_ok and download_ok) else 'PARTIAL FAILURE'}")
print("=" * 70)
```

```python
"""
CELL 3 v7b — RECALIBRATED CALIBRATION + WIDTH SWEEP
=====================================================

Two changes from v7:

CHANGE 1: recalibrate S_M so that alpha_fit = 2G/c^2 exactly.
  From v7: alpha_fit = 1.586e6 m/kg.
  Target:  alpha = 2G/c^2 = 1.485e-27 m/kg.
  New S_M = S_M_old * (alpha_fit / alpha_target)
          = 1e-8 * (1.586e6 / 1.485e-27)
          = 1e-8 * 1.068e33
          = 1.068e25 kg/unit.

CHANGE 2: width sweep instead of amplitude sweep.
  Instead of varying A, vary the Gaussian width sigma.
  If R_peak ~ M_Pi holds under width change (not just amplitude change),
  the proportionality is structural, not trivial.

Run both: recalibrated amplitude sweep AND width sweep.
"""

import os
import shutil
import json
import numpy as np
from datetime import datetime


try:
    from google.colab import drive, files
    IN_COLAB = True
    drive.mount('/content/drive')
except ImportError:
    IN_COLAB = False
    print("Not in Colab. Drive and download steps will be simulated.")


PROJECT_NAME = "FRCMD_HORIZON_HARNESS_V7B"
timestamp    = datetime.now().strftime("%Y%m%d_%H%M%S")
output_dir   = f"output_{timestamp}"
os.makedirs(output_dir, exist_ok=True)
print(f"Workspace: {output_dir}/")


# =============================================================================
# CALIBRATION (RECALIBRATED)
# =============================================================================
S_X = 1.0e-35            # m per lattice unit (unchanged)
S_M = 1.068e25           # kg per unit-load  [RECALIBRATED]
G_CLASSICAL = 6.67430e-11
C_CLASSICAL = 299792458.0


# =============================================================================
# FROZEN CORE CONSTANTS
# =============================================================================
PHI       = 1.618033988749895
PI_MAX    = 5.9259
KO_SIGMA  = 0.045

ALPHA_HYB = 1.0
BETA_HYB  = 0.1
GAMMA_HYB = 0.1
I_G       = 1.0

IK_COEFF  = 1.0 + 1.01
I2_PIVOT  = PI_MAX
IK_PIVOT  = PI_MAX

FIELD_CLIP = 1.0e19

CFL_FACTOR = 0.4
DT_MIN     = 1.0e-12
N_DEFAULT  = 256
L_DEFAULT  = 20.0
DT_MAX     = 0.005 * (L_DEFAULT / N_DEFAULT) ** 2


# =============================================================================
# OPERATORS
# =============================================================================
def dPhi_hyb_dP_yx(P_yx, I1):
    g = I1**2 / (I1**2 + I_G**2)
    abs_p  = np.abs(P_yx)
    sign_p = np.sign(P_yx)
    num = 2.0 * P_yx * (1.0 + GAMMA_HYB * abs_p) - GAMMA_HYB * sign_p * P_yx**2
    den = (1.0 + GAMMA_HYB * abs_p) ** 2
    return ALPHA_HYB + g * BETA_HYB * num / den


def _KO(f, dx):
    stencil = (np.roll(f, -2) - 4.0*np.roll(f, -1) + 6.0*f
               - 4.0*np.roll(f, 1) + np.roll(f, 2))
    return KO_SIGMA * stencil / (dx**4)


def operator_v6(P_xx, P_xy, P_yx, P_yy, dx):
    P_xx_safe = np.clip(P_xx, -FIELD_CLIP, FIELD_CLIP)
    P_yy_safe = np.clip(P_yy, -FIELD_CLIP, FIELD_CLIP)
    P_xy_safe = np.clip(P_xy, -FIELD_CLIP, FIELD_CLIP)
    P_yx_safe = np.clip(P_yx, -FIELD_CLIP, FIELD_CLIP)

    I1 = P_xx_safe + P_yy_safe
    I2 = (P_xx_safe**2 + P_xy_safe**2
          + P_yx_safe**2 + P_yy_safe**2)
    I4 = P_xx_safe**4 + P_yy_safe**4
    I1_safe = np.clip(I1, 1e-3, None)

    def D(f):
        return (np.roll(f, -1) - np.roll(f, 1)) / (2.0 * dx)

    exp_arg = -0.5 * (I2**2 + I4**4)
    exp_arg = np.clip(exp_arg, -50.0, 0.0)
    envelope = np.exp(exp_arg)

    sat2 = 1.0 + (I2 ** 2) / (I2_PIVOT ** 2)

    def C_term(P_comp_safe, P_comp_raw, D_P):
        I_k = IK_COEFF * P_comp_safe
        I_k_sq = I_k ** 2
        sat3 = 1.0 + I_k_sq / (IK_PIVOT ** 2)
        term1 = 0.2 * (D_P * I1)
        term2 = 0.2 * (I2 - I1) * (I1 + I2) / sat2
        term3 = 0.1 * I_k_sq / sat3
        term4 = ((1.0 / PI_MAX) * (I1_safe**(-0.5) - 1.0)
                 * envelope * P_comp_raw)
        return term1 + term2 + term3 + term4

    dPxx = C_term(P_xx_safe, P_xx, D(P_xx_safe))
    dPxy = C_term(P_xy_safe, P_xy, D(P_xy_safe))
    dPyx = C_term(P_yx_safe, P_yx, D(P_yx_safe))
    dPyy = C_term(P_yy_safe, P_yy, D(P_yy_safe))
    dPyx = dPyx + dPhi_hyb_dP_yx(P_yx_safe, I1)

    return (dPxx - _KO(P_xx_safe, dx),
            dPxy - _KO(P_xy_safe, dx),
            dPyx - _KO(P_yx_safe, dx),
            dPyy - _KO(P_yy_safe, dx))


# =============================================================================
# SINGLE RUN WITH VARIABLE WIDTH
# =============================================================================
def run_single(A=1.0, sigma=1.0, steps=20000,
               N=N_DEFAULT, L=L_DEFAULT, sample_every=1000,
               verbose=False):
    """
    Run v6 with initial amplitude A and Gaussian width sigma.
    """
    x  = np.linspace(-L/2, L/2, N, endpoint=False)
    dx = L / N

    P_xx = PHI + A * np.exp(-((x + 3.0) / sigma)**2)
    P_yy = PHI + A * np.exp(-((x - 3.0) / sigma)**2)
    P_xy = np.zeros(N)
    P_yx = np.zeros(N)

    dt = DT_MAX
    log = []

    for step in range(steps + 1):
        dPxx, dPxy, dPyx, dPyy = operator_v6(P_xx, P_xy, P_yx, P_yy, dx)

        D_Pxx = (np.roll(P_xx, -1) - np.roll(P_xx, 1)) / (2.0 * dx)
        max_grad = float(np.max(np.abs(D_Pxx)))

        dt_new = CFL_FACTOR * dx / (1.0 + max_grad / PI_MAX)
        dt_new = max(DT_MIN, min(DT_MAX, dt_new))
        dt = dt_new

        P_xx = P_xx + dt * dPxx
        P_xy = P_xy + dt * dPxy
        P_yx = P_yx + dt * dPyx
        P_yy = P_yy + dt * dPyy

        if step % sample_every == 0 or step == steps:
            rho_pi = np.abs(P_xx) + np.abs(P_yy)
            M_Pi = S_M * np.sum(rho_pi) * (dx * S_X)
            R_s_target = 2.0 * G_CLASSICAL * M_Pi / (C_CLASSICAL ** 2)

            r_phys = np.abs(x) * S_X
            r_safe = np.maximum(r_phys, 1e-40)
            prefactor = 2.0 * G_CLASSICAL * M_Pi / (C_CLASSICAL ** 2)
            C_sq_effective = 1.0 - (prefactor / r_safe)
            min_C_sq = float(np.min(C_sq_effective))

            peak_idx = int(np.argmax(np.abs(P_xx)))
            R_peak = float(np.abs(x[peak_idx]) * S_X)

            log.append({
                "step":         step,
                "A":            float(A),
                "sigma":        float(sigma),
                "M_Pi_kg":      float(M_Pi),
                "R_s_target_m": float(R_s_target),
                "R_peak_m":     float(R_peak),
                "min_C_sq":     float(min_C_sq),
                "max_grad":     float(max_grad),
                "max_Pxx":      float(np.max(np.abs(P_xx))),
            })

        finite = (np.isfinite(P_xx).all() and np.isfinite(P_yy).all()
                  and np.isfinite(P_xy).all() and np.isfinite(P_yx).all())
        if not finite:
            if verbose:
                print(f"  A={A} sigma={sigma}  ABORT at step {step}")
            break

    if verbose and log:
        last = log[-1]
        print(f"  A={A:.3f} sig={sigma:.3f}  step={step}  "
              f"M_Pi={last['M_Pi_kg']:.4e}  R_s={last['R_s_target_m']:.4e}  "
              f"R_peak={last['R_peak_m']:.4e}  minC^2={last['min_C_sq']:.6f}")

    return log


# =============================================================================
# LINEAR FIT
# =============================================================================
def linear_fit_through_origin(M_vals, R_vals):
    denom = np.sum(M_vals ** 2)
    if denom <= 0:
        return None, None
    alpha = float(np.sum(M_vals * R_vals) / denom)
    R_fit = alpha * M_vals
    residuals = R_vals - R_fit
    rel = residuals / np.maximum(np.abs(R_vals), 1e-30)
    return alpha, float(np.max(np.abs(rel))), float(np.mean(np.abs(rel)))


# =============================================================================
# SWEEP 1 — AMPLITUDE, RECALIBRATED
# =============================================================================
print()
print("#" * 100)
print("# SWEEP 1 — AMPLITUDE, recalibrated S_M = {:.4e}".format(S_M))
print("#" * 100)

AMPLITUDES = [0.1, 0.5, 1.0, 2.0, 5.0, 10.0, 20.0, 50.0]
STEPS = 20000
SAMPLE_EVERY = 1000

summary_A = []
print(f"{'A':>8} | {'M_Pi':>13} | {'R_s_target':>13} | "
      f"{'R_peak':>13} | {'min_C^2':>11}")
print("-" * 100)
for A in AMPLITUDES:
    log = run_single(A=A, sigma=1.0, steps=STEPS, sample_every=SAMPLE_EVERY)
    if log:
        last = log[-1]
        print(f"{A:>8.3f} | {last['M_Pi_kg']:>13.4e} | "
              f"{last['R_s_target_m']:>13.4e} | "
              f"{last['R_peak_m']:>13.4e} | "
              f"{last['min_C_sq']:>11.6f}")
        summary_A.append(last)

if len(summary_A) >= 3:
    M_vals = np.array([s["M_Pi_kg"] for s in summary_A])
    R_vals = np.array([s["R_peak_m"] for s in summary_A])
    alpha, max_rel, mean_rel = linear_fit_through_origin(M_vals, R_vals)
    alpha_Schw = 2.0 * G_CLASSICAL / (C_CLASSICAL ** 2)
    print()
    print(f"  alpha_fit       = {alpha:.6e}")
    print(f"  alpha_Schw      = {alpha_Schw:.6e}")
    print(f"  ratio           = {alpha / alpha_Schw:.6e}")
    print(f"  mean_rel_resid  = {mean_rel:.6e}")

# =============================================================================
# SWEEP 2 — WIDTH, RECALIBRATED
# =============================================================================
print()
print("#" * 100)
print("# SWEEP 2 — WIDTH, A = 1.0, recalibrated S_M = {:.4e}".format(S_M))
print("#" * 100)

SIGMAS = [0.5, 0.75, 1.0, 1.5, 2.0, 3.0, 4.0, 6.0]

summary_sigma = []
print(f"{'sigma':>8} | {'M_Pi':>13} | {'R_s_target':>13} | "
      f"{'R_peak':>13} | {'min_C^2':>11}")
print("-" * 100)
for sigma in SIGMAS:
    log = run_single(A=1.0, sigma=sigma, steps=STEPS, sample_every=SAMPLE_EVERY)
    if log:
        last = log[-1]
        print(f"{sigma:>8.3f} | {last['M_Pi_kg']:>13.4e} | "
              f"{last['R_s_target_m']:>13.4e} | "
              f"{last['R_peak_m']:>13.4e} | "
              f"{last['min_C_sq']:>11.6f}")
        summary_sigma.append(last)

if len(summary_sigma) >= 3:
    M_vals = np.array([s["M_Pi_kg"] for s in summary_sigma])
    R_vals = np.array([s["R_peak_m"] for s in summary_sigma])
    alpha, max_rel, mean_rel = linear_fit_through_origin(M_vals, R_vals)
    alpha_Schw = 2.0 * G_CLASSICAL / (C_CLASSICAL ** 2)
    print()
    print(f"  alpha_fit       = {alpha:.6e}")
    print(f"  alpha_Schw      = {alpha_Schw:.6e}")
    print(f"  ratio           = {alpha / alpha_Schw:.6e}")
    print(f"  mean_rel_resid  = {mean_rel:.6e}")

# =============================================================================
# WRITE LOGS
# =============================================================================
with open(os.path.join(output_dir, "sweep_amplitude.json"), "w") as f:
    json.dump(summary_A, f, indent=2)
with open(os.path.join(output_dir, "sweep_width.json"), "w") as f:
    json.dump(summary_sigma, f, indent=2)
print()
print(f"Wrote {output_dir}/sweep_amplitude.json  ({len(summary_A)} entries)")
print(f"Wrote {output_dir}/sweep_width.json      ({len(summary_sigma)} entries)")


# =============================================================================
# PRESERVATION
# =============================================================================
zip_base = f"{PROJECT_NAME}_{timestamp}"
zip_path = shutil.make_archive(zip_base, 'zip', output_dir)

drive_ok = False
drive_dir = f"/content/drive/MyDrive/{PROJECT_NAME}/{output_dir}"
if IN_COLAB:
    try:
        os.makedirs(drive_dir, exist_ok=True)
        for item in os.listdir(output_dir):
            shutil.copy2(os.path.join(output_dir, item),
                         os.path.join(drive_dir, item))
        shutil.copy2(zip_path, f"/content/drive/MyDrive/{PROJECT_NAME}/{zip_base}.zip")
        drive_ok = True
    except Exception as e:
        print(f"Drive backup failed: {e}")

download_ok = False
if IN_COLAB:
    try:
        files.download(zip_path)
        download_ok = True
    except Exception as e:
        print(f"Download failed: {e}")


print()
print(f"{'✓' if os.path.exists(output_dir) else '✗'} Colab workspace saved")
print(f"{'✓' if drive_ok else '✗'} Google Drive backup saved")
print(f"{'✓' if download_ok else '✗'} Download package created")
print()
print("=" * 70)
print(" HORIZON DERIVATION HARNESS v7b — BENCH READOUT")
print("=" * 70)
print(f" OUTPUT DIRECTORY: {os.path.abspath(output_dir)}")
print(f" MASTER ZIP: {os.path.abspath(zip_path)}")
print(f" S_M: {S_M:.4e}")
print(f" STATUS: {'SUCCESS' if (drive_ok and download_ok) else 'PARTIAL FAILURE'}")
print("=" * 70)
```

```python
# =============================================================================
# CONFIG A — v4 as written (saturating denom, thresholds = PI_MAX)
# =============================================================================
def operator_A(P_xx, P_xy, P_yx, P_yy, dx):
    I1 = P_xx + P_yy
    I2 = P_xx**2 + P_xy**2 + P_yx**2 + P_yy**2
    I1_safe = np.clip(I1, 1e-3, None)

    def D(f):
        return (np.roll(f, -1) - np.roll(f, 1)) / (2.0 * dx)

    def C_term(P_comp, D_P):
        I_k = IK_COEFF * P_comp
        term1 = 0.2 * (D_P * I1)
        # Saturating denominator scaled to PI_MAX
        term2 = 0.2 * (I2 - I1) * (I1 + I2) / (1.0 + (I2 / PI_MAX)**2)
        term3 = 0.1 * (I_k**2) / (1.0 + (np.abs(I_k) / PI_MAX)**2)
        term4 = ((1.0 / PI_MAX) * (I1_safe**(-0.5) - 1.0) * P_comp)
        return term1 + term2 + term3 + term4

    dPxx = C_term(P_xx, D(P_xx)) - _KO(P_xx, dx)
    dPxy = C_term(P_xy, D(P_xy)) - _KO(P_xy, dx)
    dPyx = C_term(P_yx, D(P_yx)) - _KO(P_yx, dx) + dPhi_hyb_dP_yx(P_yx, I1)
    dPyy = C_term(P_yy, D(P_yy)) - _KO(P_yy, dx)

    return dPxx, dPxy, dPyx, dPyy


# =============================================================================
# CONFIG B — v4 with higher thresholds (I2_PIVOT = 100, IK_PIVOT = 100)
# =============================================================================
def operator_B(P_xx, P_xy, P_yx, P_yy, dx):
    I1 = P_xx + P_yy
    I2 = P_xx**2 + P_xy**2 + P_yx**2 + P_yy**2
    I1_safe = np.clip(I1, 1e-3, None)

    HIGH_PIVOT = 100.0

    def D(f):
        return (np.roll(f, -1) - np.roll(f, 1)) / (2.0 * dx)

    def C_term(P_comp, D_P):
        I_k = IK_COEFF * P_comp
        term1 = 0.2 * (D_P * I1)
        term2 = 0.2 * (I2 - I1) * (I1 + I2) / (1.0 + (I2 / HIGH_PIVOT)**2)
        term3 = 0.1 * (I_k**2) / (1.0 + (np.abs(I_k) / HIGH_PIVOT)**2)
        term4 = ((1.0 / PI_MAX) * (I1_safe**(-0.5) - 1.0) * P_comp)
        return term1 + term2 + term3 + term4

    dPxx = C_term(P_xx, D(P_xx)) - _KO(P_xx, dx)
    dPxy = C_term(P_xy, D(P_xy)) - _KO(P_xy, dx)
    dPyx = C_term(P_yx, D(P_yx)) - _KO(P_yx, dx) + dPhi_hyb_dP_yx(P_yx, I1)
    dPyy = C_term(P_yy, D(P_yy)) - _KO(P_yy, dx)

    return dPxx, dPxy, dPyx, dPyy


# =============================================================================
# CONFIG C — v3 baseline (no saturation)
# =============================================================================
def operator_C(P_xx, P_xy, P_yx, P_yy, dx):
    I1 = P_xx + P_yy
    I2 = P_xx**2 + P_xy**2 + P_yx**2 + P_yy**2
    I1_safe = np.clip(I1, 1e-3, None)

    def D(f):
        return (np.roll(f, -1) - np.roll(f, 1)) / (2.0 * dx)

    def C_term(P_comp, D_P):
        I_k = IK_COEFF * P_comp
        term1 = 0.2 * (D_P * I1)
        term2 = 0.2 * (I2 - I1) * (I1 + I2)
        term3 = 0.1 * (I_k**2)
        term4 = ((1.0 / PI_MAX) * (I1_safe**(-0.5) - 1.0) * P_comp)
        return term1 + term2 + term3 + term4

    dPxx = C_term(P_xx, D(P_xx)) - _KO(P_xx, dx)
    dPxy = C_term(P_xy, D(P_xy)) - _KO(P_xy, dx)
    dPyx = C_term(P_yx, D(P_yx)) - _KO(P_yx, dx) + dPhi_hyb_dP_yx(P_yx, I1)
    dPyy = C_term(P_yy, D(P_yy)) - _KO(P_yy, dx)

    return dPxx, dPxy, dPyx, dPyy


# =============================================================================
# CONFIG D — v4 + explicit fixed-point pull
# =============================================================================
def operator_D(P_xx, P_xy, P_yx, P_yy, dx):
    I1 = P_xx + P_yy
    I2 = P_xx**2 + P_xy**2 + P_yx**2 + P_yy**2
    I1_safe = np.clip(I1, 1e-3, None)

    def D(f):
        return (np.roll(f, -1) - np.roll(f, 1)) / (2.0 * dx)

    def C_term(P_comp, D_P):
        I_k = IK_COEFF * P_comp
        term1 = 0.2 * (D_P * I1)
        term2 = 0.2 * (I2 - I1) * (I1 + I2) / (1.0 + (I2 / PI_MAX)**2)
        term3 = 0.1 * (I_k**2) / (1.0 + (np.abs(I_k) / PI_MAX)**2)
        term4 = ((1.0 / PI_MAX) * (I1_safe**(-0.5) - 1.0) * P_comp)

        # Explicit pull toward I2 = PI_MAX^2
        pull = -0.05 * (I2 - PI_MAX**2) * P_comp

        return term1 + term2 + term3 + term4 + pull

    dPxx = C_term(P_xx, D(P_xx)) - _KO(P_xx, dx)
    dPxy = C_term(P_xy, D(P_xy)) - _KO(P_xy, dx)
    dPyx = C_term(P_yx, D(P_yx)) - _KO(P_yx, dx) + dPhi_hyb_dP_yx(P_yx, I1)
    dPyy = C_term(P_yy, D(P_yy)) - _KO(P_yy, dx)

    return dPxx, dPxy, dPyx, dPyy


# =============================================================================
# INTEGRATION AND PRESERVATION HARNESS
# =============================================================================
import os
import json
import shutil
from datetime import datetime

try:
    from google.colab import files, drive
    IN_COLAB = True
    drive.mount('/content/drive', force_remount=False)
except ImportError:
    IN_COLAB = False

def run_test(operator_func, label, N=256, L=20.0, steps=5000, dt_coeff=0.005):
    x  = np.linspace(-L/2, L/2, N, endpoint=False)
    dx = L / N
    dt = dt_coeff * (dx ** 2)

    P_xx = PHI + np.exp(-(x + 3.0)**2)
    P_yy = PHI + np.exp(-(x - 3.0)**2)
    P_xy = np.zeros(N)
    P_yx = np.zeros(N)

    print(f"\n--- RUNNING {label} ---")

    for step in range(steps + 1):
        dPxx, dPxy, dPyx, dPyy = operator_func(P_xx, P_xy, P_yx, P_yy, dx)

        P_xx = P_xx + dt * dPxx
        P_xy = P_xy + dt * dPxy
        P_yx = P_yx + dt * dPyx
        P_yy = P_yy + dt * dPyy

        finite = (np.isfinite(P_xx).all() and np.isfinite(P_yy).all()
                  and np.isfinite(P_xy).all() and np.isfinite(P_yx).all())

        if not finite:
            print(f"CRITICAL: {label} DIVERGED at step {step}.")
            return {"label": label, "status": "diverged", "step": step}

    print(f"SUCCESS: {label} CONVERGED through {steps} steps.")
    return {"label": label, "status": "converged", "step": steps}


# Execute Matrix
PROJECT_NAME = "FRCMD_ASYMPTOTE_TEST"
timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
output_dir = f"output_{timestamp}"
os.makedirs(output_dir, exist_ok=True)

results = []
results.append(run_test(operator_A, "CONFIG A (v4 standard)"))
results.append(run_test(operator_B, "CONFIG B (v4 high threshold)"))
results.append(run_test(operator_C, "CONFIG C (v3 baseline)"))
results.append(run_test(operator_D, "CONFIG D (v4 + fixed point pull)"))

# STEP 1: SAVE TO COLAB WORKSPACE
output_json_path = os.path.join(output_dir, "asymptote_results.json")
with open(output_json_path, "w") as f:
    json.dump(results, f, indent=4)
print("\n✓ Colab workspace saved")

# STEP 2: CREATE MASTER ZIP
zip_name = f"{PROJECT_NAME}_{timestamp}"
zip_path = shutil.make_archive(zip_name, 'zip', output_dir)
zip_full_path = f"{zip_name}.zip"

# STEP 3: BACKUP TO GOOGLE DRIVE
drive_backup_path = f"/content/drive/MyDrive/{PROJECT_NAME}/"
drive_status = False
if IN_COLAB:
    try:
        os.makedirs(drive_backup_path, exist_ok=True)
        shutil.copytree(output_dir, os.path.join(drive_backup_path, output_dir), dirs_exist_ok=True)
        shutil.copy2(zip_full_path, drive_backup_path)
        drive_status = True
        print("✓ Google Drive backup saved")
    except Exception as e:
        print(f"Drive backup failed: {e}")
else:
    print("✓ Google Drive backup saved (Simulated)")
    drive_status = True

# STEP 4: DOWNLOAD TO LOCAL MACHINE
download_status = False
if IN_COLAB:
    try:
        files.download(zip_full_path)
        download_status = True
        print("✓ Download package created")
    except Exception as e:
        print(f"Download trigger failed: {e}")
else:
    print("✓ Download package created (Simulated)")
    download_status = True

# STEP 5 & 6: VERIFY FILES EXIST & FINAL STATUS REPORT
file_count = len(os.listdir(output_dir))
archive_size = os.path.getsize(zip_full_path)
workspace_exists = os.path.exists(output_dir)
zip_exists = os.path.exists(zip_full_path)

if workspace_exists and zip_exists and drive_status and download_status:
    final_status = "SUCCESS"
else:
    final_status = "FAILED - PARTIAL PRESERVATION DETECTED"

print("\n" + "="*70)
print(" FINAL STATUS REPORT")
print("="*70)
print(f" OUTPUT DIRECTORY: {os.path.abspath(output_dir)}")
print(f" GOOGLE DRIVE BACKUP: {drive_backup_path}")
print(f" MASTER ZIP: {os.path.abspath(zip_full_path)}")
print(f" FILE COUNT: {file_count}")
print(f" ARCHIVE SIZE: {archive_size}")
print(f" STATUS: {final_status}")
print("="*70)
```
